## Question 1 | Boolean Expression (8 pts)

Evaluate each expression. / 求出每个表达式的值。

### Answers | 答案

| # | Expression | Value |
|---|---|---|
| 1 | `'ace' in 'space' or 5%2 < 1` | **True** |
| 2 | `3 == '3' or 'pen' > 'pencil'` | **False** |
| 3 | `7//3 > 2 or len('abs') == 3 or not '12' == 12` | **True** |
| 4 | `not True and True and True` | **False** |

### How to think | 答题思路

> **EN** Use three steps. One: evaluate each small piece. Two: apply operator precedence from the cheat sheet: arithmetic first, then comparisons and `in`, then `not`, then `and`, and `or` last. Three: remember short-circuiting: `or` is True as soon as one side is True, and `and` is False as soon as one side is False.
>
> **中** 分三步：第一，先算每个小部分；第二，按小抄上的优先级组合：先算术，再比较和 `in`，然后 `not`，再 `and`，最后 `or`；第三，记住短路：`or` 只要有一边为 True 就是 True，`and` 只要有一边为 False 就是 False。

**(1)** `'ace' in 'space' or 5%2 < 1` → **True**

> **EN** `in` checks whether one string appears inside another. "space" is s-p-**a-c-e**, so `'ace' in 'space'` is True. The left side of `or` is already True, so the whole expression is True. (The right side, `5%2 < 1`, is `1 < 1`, which is False, but it doesn't matter.)
>
> **中** `in` 判断一个字符串是否出现在另一个字符串里。"space" 是 s-p-**a-c-e**，所以 `'ace' in 'space'` 为 True。`or` 左边已经是 True，整个表达式就是 True。（右边 `5%2 < 1` 是 `1 < 1`，为 False，但已经不影响结果。）

**(2)** `3 == '3' or 'pen' > 'pencil'` → **False**

> **EN** `3` is an integer and `'3'` is a string. A number and a string are never equal, so `3 == '3'` is False. Next, strings are compared character by character. The first three characters, p-e-n, are the same, and then `'pen'` runs out. When one string is the beginning of the other, the shorter one is smaller. So `'pen' < 'pencil'`, and `'pen' > 'pencil'` is False. False or False gives False.
>
> **中** `3` 是整数，`'3'` 是字符串，数字和字符串永远不相等，所以 `3 == '3'` 为 False。字符串比较是逐个字符比：前三个字符 p-e-n 都一样，然后 `'pen'` 没有字符了。当一个字符串是另一个的开头部分时，较短的那个更小，所以 `'pen' < 'pencil'`，`'pen' > 'pencil'` 为 False。False or False 得到 False。

**(3)** `7//3 > 2 or len('abs') == 3 or not '12' == 12` → **True**

> **EN** `7//3` is floor division, so it is 2, not 2.33. `2 > 2` is False. Then `len('abs')` is 3, so `len('abs') == 3` is True, and the whole `or` chain is True. For completeness, `not '12' == 12` means `not ('12' == 12)`, because `==` is evaluated before `not`. That is `not False`, which is True.
>
> **中** `7//3` 是整除，结果是 2，不是 2.33。`2 > 2` 为 False。接着 `len('abs')` 是 3，`len('abs') == 3` 为 True，整个 `or` 链就是 True。补充：`not '12' == 12` 的意思是 `not ('12' == 12)`，因为 `==` 比 `not` 先算，也就是 `not False`，为 True。

**(4)** `not True and True and True` → **False**

> **EN** `not` is evaluated before `and`, so this is `(not True) and True and True`, which is `False and True and True`. An `and` with a False is False. Reading it as `not (True and True and True)` would also give False here, but the correct reading is `(not True) and ...`.
>
> **中** `not` 比 `and` 先算，所以这是 `(not True) and True and True`，也就是 `False and True and True`。只要 `and` 里有一个 False，结果就是 False。如果理解成 `not (True and True and True)`，这里碰巧也是 False，但正确的读法是 `(not True) and ...`。

### ⚠️ Pitfalls | 易错点

- 以为 `3 == '3'` 为 True：整数和字符串永远不相等。/ Thinking `3 == '3'` is True.
- 以为长的字符串一定"更大"，或者按字母个数比较：实际是逐字符比较，前缀更短的那个更小。/ Comparing strings by length instead of character by character.
- 把 `7//3` 算成 2.33：`//` 是整除。/ Treating `//` like `/`.

````python
# Print each expression to check the answers live.
# 逐个打印四个表达式的值，现场核对答案。
print('ace' in 'space' or 5%2 < 1)
print(3 == '3' or 'pen' > 'pencil')
print(7//3 > 2 or len('abs') == 3 or not '12' == 12)
print(not True and True and True)
````

````python
# Break each expression into smaller pieces.
# 把每个表达式拆成小块逐一检查。
print('ace' in 'space', 5 % 2, 5 % 2 < 1)
print(3 == '3', 'pen' > 'pencil')
print(7 // 3, 7 // 3 > 2, len('abs') == 3, '12' == 12, not '12' == 12)
print(not True, (not True) and True and True)
````

## Question 2 | Binary and Hexadecimal Representation (8 pts)

Convert each number into the other two number systems. / 把每个数转换成另外两种进制。

### Answers | 答案

| Decimal | Binary | Hexadecimal |
|---:|---:|---:|
| **43** | `0b101011` (given) | **`0x2B`** |
| **90** | **`0b1011010`** | `0x5A` (given) |

（`0x2b` 与 `0x2B` 相同，Python 打印小写；`0b01011010` 与 `0b1011010` 数值相同。/ Lowercase and leading zeros do not change the value.）

### How to think | 答题思路

**Row 1: binary → decimal / 二进制 → 十进制**

```text
place value:  32  16   8   4   2   1
bit:           1   0   1   0   1   1
              32 +     8 +     2 + 1  = 43
```

> **EN** Each binary position is worth twice the one to its right: 1, 2, 4, 8, 16, 32 from the right. Add the place values where the bit is 1: 32 + 8 + 2 + 1 = 43.
>
> **中** 二进制每一位的权重是右边一位的两倍，从右往左依次是 1、2、4、8、16、32。把是 1 的那些位的权重加起来：32 + 8 + 2 + 1 = 43。

**Row 1: binary → hexadecimal / 二进制 → 十六进制**

```text
101011  →  10 | 1011  →  0010 | 1011  →  2 | B  →  0x2B
```

> **EN** One hex digit is exactly 4 bits. Group the bits in fours **starting from the right**, pad the left group with zeros, and convert each group: 0010 is 2 and 1011 is 11, which is B. So the answer is `0x2B`.
>
> **中** 一个十六进制位正好对应 4 个二进制位。**从右往左**每 4 位一组，最左边不够的补 0，再逐组转换：0010 是 2，1011 是 11，也就是 B。所以答案是 `0x2B`。

**Row 2: hexadecimal → decimal / 十六进制 → 十进制**

> **EN** In hex, A means 10, B means 11, up to F, which means 15. `0x5A` is 5 × 16 + 10 = 80 + 10 = 90. You can also check this on the cheat sheet: `hex(90)` gives `0x5A`.
>
> **中** 十六进制中 A 表示 10，B 表示 11……F 表示 15。`0x5A` = 5 × 16 + 10 = 80 + 10 = 90。小抄上 `hex(90)` 的结果正好就是 `0x5A`，可以用来核对。

**Row 2: hexadecimal → binary / 十六进制 → 二进制**

```text
0x5A  →  5 | A  →  0101 | 1010  →  0b1011010
```

> **EN** Turn each hex digit into 4 bits: 5 is 0101 and A (10) is 1010. Together that's 0101 1010; dropping the leading zero gives `0b1011010`. Check: 64 + 16 + 8 + 2 = 90.
>
> **中** 每个十六进制位换成 4 个二进制位：5 是 0101，A（10）是 1010。合起来是 0101 1010，去掉开头的 0 就是 `0b1011010`。核对：64 + 16 + 8 + 2 = 90。

| Hex digit | 0–9 | A | B | C | D | E | F |
|---|---|---|---|---|---|---|---|
| Value | 0–9 | 10 | 11 | 12 | 13 | 14 | 15 |

### ⚠️ Pitfalls | 易错点

- 从左边开始每 4 位分组（`1010 | 11`），得到错误的 `0xA3`。/ Grouping bits from the left gives the wrong `0xA3`.
- 忘记 A = 10，把 `0x5A` 算成 5 × 16 + 1。/ Forgetting that A means 10.
- 二进制的位权从右边的 1 开始，不是从左边开始。/ Place values start at 1 on the right.

````python
# A literal written with 0b (binary) or 0x (hex) is just a number to Python.
# 以 0b（二进制）或 0x（十六进制）开头的写法，在 Python 里就是一个数。
print(0b101011)   # 43
# hex() and bin() turn a number into hex/binary text (see the cheat sheet).
# Python prints hex digits in lowercase: 0x2b is the same as 0x2B.
# hex() 和 bin() 把数转成十六进制 / 二进制文本（见小抄）；Python 用小写，0x2b 与 0x2B 相同。
print(hex(43))    # 0x2b
print(0x5A)       # 90
print(bin(90))    # 0b1011010
````

### Program 4 — `break` at the top of the loop

```python
count = 0

for i in range(1, 13):
    if count > 4:
        break

    if i % 2 == 0 or i % 3 == 0:
        print(i)
        count += 1

print("End")
```

**Output**

```text
2
3
4
6
8
End
```

| `i` | `count` at start | `count > 4`? | even or multiple of 3? | printed | `count` after |
|---:|---:|---|---|---|---:|
| 1 | 0 | no | no | — | 0 |
| 2 | 0 | no | yes | 2 | 1 |
| 3 | 1 | no | yes | 3 | 2 |
| 4 | 2 | no | yes | 4 | 3 |
| 5 | 3 | no | no | — | 3 |
| 6 | 3 | no | yes | 6 | 4 |
| 7 | 4 | no | no | — | 4 |
| 8 | 4 | no | yes | 8 | 5 |
| 9 | 5 | **yes → `break`** | — | — | 5 |

> **EN** The program prints numbers that are even **or** multiples of 3. The key detail is **where** the `break` check is: at the **top** of the loop. After 8 is printed, `count` becomes 5, but the loop doesn't stop yet. It stops at the start of the next round, when `i` is 9. So 9 is never printed, even though 9 is a multiple of 3. Finally, `End` is printed after the loop.
>
> **中** 程序打印"偶数**或** 3 的倍数"。关键在于 `break` 的检查放在循环体的**开头**：打印 8 之后 `count` 变成 5，但循环不会马上停，而是在下一轮开头（`i` 为 9 时）才停。所以 9 虽然是 3 的倍数，也不会被打印。最后在循环外打印 `End`。

⚠️ **Pitfalls**

- 把 9 也打印出来，忽略了下一轮开头的 `break`。/ Printing 9 as well.
- 以为 `count` 到 4 就停，只写出 4 个数：`count > 4` 要等 `count` 为 5 才成立。/ Stopping after four numbers: `count > 4` needs `count` to be 5.
- 6 同时满足两个条件，但只打印一次：`or` 只要一边为真即可。/ 6 satisfies both conditions but is printed once.

````python
# Program 4, exactly as in the exam.
# 原卷 Program 4。
count = 0

for i in range(1, 13):
    if count > 4:
        break

    if i % 2 == 0 or i % 3 == 0:
        print(i)
        count += 1

print("End")
````

### How to think | 答题思路

**Step 1 — give every variable a job / 先搞清楚每个变量的作用**

> **EN** This is the same skeleton as Exercise 1 of Recitation 04/05. `i` marks the **start of a group** of equal neighbouring letters. `j` is a scout. It starts at `i + 1` and moves right while it sees the same letter. When the inner loop stops, `j` is the start of the next group, and `i = j` jumps there.
>
> **中** 这和 Recitation 04/05 第 1 题是同一个框架。`i` 标记一组"相邻且相同"字母的**开头**；`j` 是探路的指针，从 `i + 1` 出发，只要看到相同的字母就继续右移。内层循环停下时，`j` 就是下一组的开头，`i = j` 直接跳过去。

**Step 2 — what does `j != i + 1` mean? / `j != i + 1` 是什么意思？**

> **EN** If the group has only one letter, `j` never moves, so `j == i + 1`. If `j != i + 1`, then `j` moved at least once, so the group has **two or more** letters. Therefore `a` counts **how many groups have length at least 2**, that is, how many groups of repeated letters there are.
>
> **中** 如果这一组只有一个字母，`j` 一步都没动，`j == i + 1`。如果 `j != i + 1`，说明 `j` 至少动了一次，这一组有**两个或更多**字母。所以 `a` 统计的是**长度至少为 2 的组有几个**，也就是"连续重复的字母组"有几个。

**Step 3 — split each string into groups / 把每个字符串分组**

```text
'abc'                →  a | b | c                              →  0 groups of length ≥ 2
'aa'                 →  aa                                     →  1
'aabbcc'             →  aa | bb | cc                           →  3
'aaaabccdeefffffgg'  →  aaaa | b | cc | d | ee | fffff | gg    →  5  (aaaa, cc, ee, fffff, gg)
```

Trace for `'aabbcc'` / 追踪 `'aabbcc'`：

| `i` | `N[i]` | `j` after the inner loop | `j != i + 1`? | `a` |
|---:|---|---:|---|---:|
| 0 | `a` | 2 | yes | 1 |
| 2 | `b` | 4 | yes | 2 |
| 4 | `c` | 6 | yes | 3 |

### ⚠️ Pitfalls | 易错点

- 数所有的组（最后一个会得到 7），而不是只数长度 ≥ 2 的组。/ Counting all groups (7 for the last string).
- 数重复的字母个数，例如把 `fffff` 算成 4 或 5。每一组只加一次。/ Counting repeated letters instead of groups.

````python
# Run the Question 4 program for each value in the table.
# (This checking loop uses a list only for verification; lists were not allowed in the exam answers.)
# 对表格中的每个值运行第 4 题的程序。
# （这里用列表只是为了验证；考试作答时不允许使用列表。）
for aString in ['abc', 'aa', 'aabbcc', 'aaaabccdeefffffgg']:
    N = aString

    i, j = 0, 0
    a = 0

    while i < len(N):

        j = i + 1
        while j < len(N) and N[j] == N[i]:
            j = j + 1

        if j != i + 1:
            a = a + 1

        i = j

    print(aString, "->", a)
````

> **EN** Read one character. If it is a letter, print whether it is uppercase or lowercase, and whether it is a vowel or a consonant. If it is a digit, print whether it is even or odd. Otherwise print "Other character".
>
> **中** 读入一个字符。如果是字母，输出大写还是小写、元音还是辅音；如果是数字，输出奇数还是偶数；否则输出 "Other character"。

Examples from the exam / 原卷样例：`E` → `Uppercase vowel`，`b` → `Lowercase consonant`，`7` → `Odd digit`，`#` → `Other character`.

### How to think | 答题思路

**Step 1 — read the input as text / 读入的是文本**

> **EN** `input()` always gives a string. Do **not** convert it with `int()` at the start: if the user types a letter, `int('E')` crashes with a `ValueError`. We convert only after we know it is a digit.
>
> **中** `input()` 返回的永远是字符串。**不要**一开始就用 `int()` 转换：用户输入字母时，`int('E')` 会报 `ValueError`。等确认是数字之后再转换。

**Step 2 — decide the category with chained comparisons / 用连续比较判断类别**

> **EN** Follow the hint: `'A' <= ch <= 'Z'` is True exactly for uppercase letters, because letters have consecutive code numbers. In the same way, `'a' <= ch <= 'z'` checks lowercase letters and `'0' <= ch <= '9'` checks digits. Anything else is "Other character". An `if`/`elif`/`elif`/`else` chain gives us exactly these four cases.
>
> **中** 按提示：`'A' <= ch <= 'Z'` 恰好对大写字母为真，因为字母的编号是连续的。同理，`'a' <= ch <= 'z'` 判断小写字母，`'0' <= ch <= '9'` 判断数字，其余都是 "Other character"。用 `if`/`elif`/`elif`/`else` 正好分成这四种情况。

**Step 3 — vowel or consonant / 元音还是辅音**

> **EN** Use the membership operator: `ch in 'AEIOU'`. `in` is an operator, not a string method, so it is allowed. Inside the uppercase branch we compare with uppercase vowels, and inside the lowercase branch with lowercase vowels.
>
> **中** 用成员运算符：`ch in 'AEIOU'`。`in` 是运算符，不是字符串方法，所以可以用。大写分支里和大写元音比较，小写分支里和小写元音比较。

**Step 4 — even or odd / 奇数还是偶数**

> **EN** Now we know `ch` is a digit, so `int(ch)` is safe. If `int(ch) % 2 == 0`, it's even; otherwise it's odd. Following the examples, print `Even digit` or `Odd digit`.
>
> **中** 这时已经确定 `ch` 是数字，`int(ch)` 是安全的。`int(ch) % 2 == 0` 为偶数，否则为奇数。按样例格式输出 `Even digit` 或 `Odd digit`。

### Answer | 参考答案

````python
# Read one character. input() returns a string, so we do not convert it yet.
# 读入一个字符。input() 返回字符串，这里先不转换。
ch = input("Enter a character: ")

# Uppercase letters: 'A' <= ch <= 'Z' means 'A' <= ch and ch <= 'Z'.
# 大写字母：'A' <= ch <= 'Z' 等价于 'A' <= ch and ch <= 'Z'。
if 'A' <= ch <= 'Z':
    # `in` checks whether ch appears in the string of vowels (an operator, not a method).
    # `in` 检查 ch 是否出现在元音字符串里（这是运算符，不是字符串方法）。
    if ch in 'AEIOU':
        print("Uppercase vowel")
    else:
        print("Uppercase consonant")
# Lowercase letters.
# 小写字母。
elif 'a' <= ch <= 'z':
    if ch in 'aeiou':
        print("Lowercase vowel")
    else:
        print("Lowercase consonant")
# Digits. Only now is int(ch) safe, because we know ch is '0' to '9'.
# 数字。只有确认 ch 在 '0' 到 '9' 之间后，int(ch) 才是安全的。
elif '0' <= ch <= '9':
    if int(ch) % 2 == 0:
        print("Even digit")
    else:
        print("Odd digit")
# Everything else: symbols, spaces, and so on.
# 其余情况：符号、空格等。
else:
    print("Other character")
````

## Question 6 | Saving Goal (10 pts)

> **EN** Ask for an initial deposit, an annual interest rate in percent, and a target. Every year the balance becomes `balance * (1 + rate / 100)`. Print the balance at the end of each year with 2 decimal places, until it reaches or exceeds the target. Then print how many years were needed.
>
> **中** 读入初始存款、年利率（百分数）和目标金额。每年余额变为 `balance * (1 + rate / 100)`。每年年末用两位小数打印余额，直到余额达到或超过目标，最后打印需要几年。

### How to think | 答题思路

**Step 1 — inputs / 输入**

> **EN** Money and interest rates can have decimals, like 3.5 percent, so we use `float(input(...))`.
>
> **中** 金额和利率可能带小数（例如 3.5%），所以用 `float(input(...))`。

**Step 2 — state / 需要记住的状态**

> **EN** We need two variables: `balance`, which starts at the deposit, and `years`, a counter that starts at 0.
>
> **中** 需要两个变量：`balance` 从初始存款开始，`years` 是从 0 开始的计数器。

**Step 3 — which loop? / 用哪种循环？**

> **EN** We don't know in advance how many years it will take, so this is a `while` loop. "Until the balance reaches or exceeds the target" means we stop when `balance >= target`. In other words, we **keep going while `balance < target`**.
>
> **中** 事先不知道要几年，所以用 `while` 循环。"直到余额达到或超过目标"表示 `balance >= target` 时停止，也就是**只要 `balance < target` 就继续**。

**Step 4 — the loop body, in the right order / 循环体的顺序**

> **EN** First update the balance with the formula, then add one to `years`, then print. If you print before updating, the first line would wrongly show the starting deposit.
>
> **中** 先用公式更新余额，再把 `years` 加一，然后打印。如果先打印再更新，第一行会错误地显示初始存款。

**Step 5 — after the loop / 循环之后**

> **EN** When the loop ends, `years` holds the answer. Print it once, outside the loop.
>
> **中** 循环结束时 `years` 就是答案，在循环外打印一次。

### Answer | 参考答案

````python
# Money and rates may have decimals, so convert with float().
# 金额和利率可能有小数，所以用 float() 转换。
deposit = float(input("Initial deposit: "))
rate = float(input("Annual interest rate (%): "))
target = float(input("Target amount: "))

# State: the current balance, and a counter for the years.
# 状态：当前余额，以及年份计数器。
balance = deposit
years = 0

# Keep going while the target is not reached yet ("until balance >= target").
# 只要还没达到目标就继续（即"直到 balance >= target"为止）。
while balance < target:
    # Apply one year of interest, using the formula from the question.
    # 按题目公式计算一年的利息。
    balance = balance * (1 + rate / 100)
    # Count this year.
    # 计入这一年。
    years += 1
    # :.2f shows exactly two decimal places.
    # :.2f 固定显示两位小数。
    print(f"Year {years}: {balance:.2f}")

# Printed once, after the loop.
# 循环结束后只打印一次。
print("Years needed:", years)
````

> **EN** `change = paid - price`. If it is 0, print the "No change" message. Otherwise use the largest piece first. `change // 10` is the number of 10-yuan pieces, and `change % 10` is what is left. Do the same with 5, and the rest is 1-yuan pieces. Print a line only when the count is more than 0.
>
> **中** `change = paid - price`。如果为 0，打印 "No change" 那句。否则从最大面额开始：`change // 10` 是 10 元的张数，`change % 10` 是剩下的金额；对 5 元做同样的事，剩下的就是 1 元的个数。只有数量大于 0 才打印那一行。

❓ **Ask** — Why do we never need a 20-yuan piece for change? / 为什么找零永远用不到 20 元？

**Answer** — Before the last payment, `paid` was still less than the price, and the last payment is at most 20, so the change is at most 19. / 最后一次投币之前 `paid` 还小于价格，而最后一次最多投 20，所以找零最多 19 元。

❓ **Ask** — Why does "largest piece first" give the fewest pieces here? / 为什么"先用大面额"就能用最少的张数？

**Answer** — 10 is made of two 5s, and 5 is made of five 1s. Replacing smaller pieces with one bigger piece always saves pieces, so we should use as many big pieces as possible. / 10 = 两个 5，5 = 五个 1。用一个大面额换掉几个小面额总能减少张数，所以大面额能用多少就用多少。

### Answer | 参考答案

````python
# Part A: the price is a positive integer.
# A 部分：价格是正整数。
price = int(input("Item price (yuan): "))
# Accumulator: the total money accepted so far.
# 累加器：目前已收下的总金额。
paid = 0

# Part B: we don't know how many coins will come, so use while.
# B 部分：不知道会投几次，所以用 while。
while paid < price:
    money = int(input("Insert money (1, 5, 10 or 20): "))
    # Lists are not allowed, so compare with each accepted value using `or`.
    # 不能用列表，所以用 `or` 和每个可接受的面额逐一比较。
    if money == 1 or money == 5 or money == 10 or money == 20:
        paid += money
        print(f"Paid: {paid} / {price}")
    else:
        # A rejected value is NOT added to paid.
        # 被拒收的金额不加到 paid 上。
        print(f"Rejected: {money}")

# Part C: change.
# C 部分：找零。
change = paid - price

if change == 0:
    print("No change. Enjoy your drink!")
else:
    print("Change:", change)
    # Largest piece first: // gives how many pieces, % gives what is left.
    # 先用最大面额：// 得到张数，% 得到剩下的金额。
    tens = change // 10
    change = change % 10
    fives = change // 5
    ones = change % 5
    # Print only the denominations that are actually used.
    # 只打印实际用到的面额。
    if tens > 0:
        print(f"10 yuan x {tens}")
    if fives > 0:
        print(f"5 yuan x {fives}")
    if ones > 0:
        print(f"1 yuan x {ones}")
    print("Enjoy your drink!")
````

### ⚠️ Pitfalls | 易错点

- **`if money == 1 or 5 or 10 or 20`**：永远为真！`5` 本身就算 True。每个比较都要写完整 `money == ...`。/ This is always True, because `5` by itself counts as True.

## Question 8 | Prime Finder (15 pts)

> **EN** Read two integers, `start` and `end`, and print all primes from `start` to `end`, inclusive, on one line separated by spaces. Then print the total count. A prime is an integer greater than 1 whose only divisors are 1 and itself.
>
> **中** 读入两个整数 `start` 和 `end`，把 `start` 到 `end`（含两端）之间的所有质数打印在同一行、用空格隔开，再打印总个数。质数是大于 1、且只能被 1 和自身整除的整数。

### How to think | 答题思路

**Step 1 — two levels of repetition / 两层重复**

> **EN** We repeat for every number in the range, and for each number we try many divisors. Two levels of repetition means nested loops. The outer loop goes through the numbers, and the inner loop tests one number.
>
> **中** 要对范围内的每个数重复，而判断每个数又要试很多个除数。两层重复就是嵌套循环：外层遍历每个数，内层判断一个数是不是质数。

**Step 2 — the outer loop / 外层循环**

> **EN** "Inclusive" means `end` must be included, so we use `range(start, end + 1)`. Remember, the stop value of `range` is not included.
>
> **中** "含两端"意味着 `end` 也要算进去，所以写 `range(start, end + 1)`；记住 `range` 不包含终点。

**Step 3 — testing one number with a flag / 用标志变量判断一个数**

> **EN** Numbers less than or equal to 1 are not prime, so we skip them. For other numbers, we start by assuming the number is prime: `is_prime = True`. Then we try every divisor from 2 to `number - 1`. As soon as one divides evenly, we set `is_prime = False` and `break`, because one divisor is enough to prove it is not prime. After the inner loop, `is_prime` tells us the answer.
>
> **中** 小于等于 1 的数不是质数，直接跳过。其他的数先假设是质数：`is_prime = True`；然后试 2 到 `number - 1` 的每个除数，一旦有一个能整除，就把 `is_prime` 设为 False 并 `break`，因为找到一个除数就足以证明它不是质数。内层循环结束后，`is_prime` 就是答案。

**Step 4 — printing on one line / 在同一行输出**

> **EN** For each prime, `print(number, end=" ")` keeps the numbers on one line, and `count += 1` counts it. After the outer loop, an empty `print()` ends that line, and then we print the total.
>
> **中** 每找到一个质数，用 `print(number, end=" ")` 让数字留在同一行，并 `count += 1` 计数。外层循环结束后，用空的 `print()` 结束这一行，再打印总数。

### Answer | 参考答案

````python
start = int(input("Start: "))
end = int(input("End: "))

# Counter for the primes found.
# 计数器：找到的质数个数。
count = 0

# "Inclusive" means end is included, so the stop value is end + 1.
# "含两端"表示要包含 end，所以终点写 end + 1。
for number in range(start, end + 1):
    # Primes are greater than 1, so 1, 0, and negative numbers are skipped.
    # 质数必须大于 1，所以 1、0 和负数直接跳过。
    if number > 1:
        # Flag: assume it is prime until a divisor proves otherwise.
        # Reset it for EVERY number.
        # 标志变量：先假设是质数，直到找到除数为止；每个数都要重新设为 True。
        is_prime = True
        for divisor in range(2, number):
            if number % divisor == 0:
                is_prime = False
                # One divisor is enough: stop testing this number.
                # 找到一个除数就够了：停止测试这个数。
                break
        if is_prime:
            # end=" " keeps all primes on one line, separated by spaces.
            # end=" " 让所有质数留在同一行，用空格隔开。
            print(number, end=" ")
            count += 1

# Finish the line of primes, then print the total.
# 先结束质数那一行，再打印总数。
print()
print("Total primes:", count)
````

```text

Total primes: 6
```


> **EN** Notice `1` in Example 2: 1 is not prime, which is why the `number > 1` check is needed. Also, `print(number, end=" ")` leaves one invisible space at the end of the line. That's normally fine. If you want to avoid it, print a space **before** every prime except the first one, the same trick as the letter triangle in Recitation 04/05.
>
> **中** 注意样例 2 中的 `1`：1 不是质数，所以需要 `number > 1` 这个判断。另外，`print(number, end=" ")` 会在行尾留下一个看不见的空格，一般没有问题；如果想避免，可以在除第一个以外的每个质数**前面**打印空格，这和 Recitation 04/05 字母三角形用的是同一个技巧。

❓ **Ask** — How can we make the inner loop faster? / 怎样让内层循环更快？

**Answer** — If a `number` greater than 1 is not prime, it has a divisor no larger than its square root. So we only need to test divisors up to the square root. The official solution does exactly this with `range(2, int(number ** 0.5) + 1)`. For exam-sized inputs, `range(2, number)` is also fine. / 如果大于 1 的 `number` 不是质数，它一定有一个不超过其平方根的除数，所以只需测试到平方根。官方答案正是这样写的：`range(2, int(number ** 0.5) + 1)`。考试规模的数据用 `range(2, number)` 也完全可以。

### ⚠️ Pitfalls | 易错点

- `range(start, end)`：漏掉 `end` 本身（例如 end 为 29 时漏掉 29）。/ Missing `end`.
- 把 1 当成质数。/ Treating 1 as prime.
- `is_prime = True` 写在外层循环之前：遇到第一个合数后，后面所有数都被判成"不是质数"。/ Setting the flag only once, outside the outer loop.
- 在内层循环里打印：一个质数会被打印多次，或者每个能整除的除数都打印一次。/ Printing inside the inner loop.
- 忘记最后的空 `print()`：`Total primes` 会接在质数那一行后面。/ Forgetting the final `print()`, so the total is on the same line.
