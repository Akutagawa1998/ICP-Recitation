## Question 1 | Boolean Expression (8 pts)

Evaluate the following boolean expression:

| # | Expression |
|---|---|
| 1 | `'ace' in 'space' or 5%2 < 1` |
| 2 | `3 == '3' or 'pen' > 'pencil'` |
| 3 | `7//3 > 2 or len('abs') == 3 or not '12' == 12` |
| 4 | `not True and True and True` |

### Answers

| # | Expression | Value |
|---|---|---|
| 1 | `'ace' in 'space' or 5%2 < 1` | **True** |
| 2 | `3 == '3' or 'pen' > 'pencil'` | **False** |
| 3 | `7//3 > 2 or len('abs') == 3 or not '12' == 12` | **True** |
| 4 | `not True and True and True` | **False** |

### Explanation

Use three steps:

1. Evaluate each small piece.
2. Apply operator precedence from the cheat sheet: arithmetic first, then comparisons and `in`, then `not`, then `and`, and `or` last.
3. Remember short-circuiting: `or` is True as soon as one side is True, and `and` is False as soon as one side is False.

**(1)** `'ace' in 'space' or 5%2 < 1` → **True**

`in` checks whether one string appears inside another. "space" is s-p-**a-c-e**, so `'ace' in 'space'` is True. The left side of `or` is already True, so the whole expression is True. The right side, `5%2 < 1`, is `1 < 1`, which is False, but that doesn't change the result.

**(2)** `3 == '3' or 'pen' > 'pencil'` → **False**

`3` is an integer and `'3'` is a string. A number and a string are never equal, so `3 == '3'` is False. Strings are compared character by character. The first three characters, p-e-n, are the same, and then `'pen'` runs out. When one string is the beginning of the other, the shorter one is smaller. So `'pen' < 'pencil'`, and `'pen' > 'pencil'` is False. False or False gives False.

**(3)** `7//3 > 2 or len('abs') == 3 or not '12' == 12` → **True**

`7//3` is floor division, so it is 2, not 2.33, and `2 > 2` is False. Next, `len('abs')` is 3, so `len('abs') == 3` is True, which makes the whole `or` chain True. For completeness, `not '12' == 12` means `not ('12' == 12)`, because `==` is evaluated before `not`. That is `not False`, which is True.

**(4)** `not True and True and True` → **False**

`not` is evaluated before `and`, so this is `(not True) and True and True`, which is `False and True and True`. An `and` with a False in it is False. Reading it as `not (True and True and True)` would also give False here, but the correct reading is `(not True) and ...`.

### Common mistakes

- Thinking `3 == '3'` is True. An integer and a string are never equal.
- Comparing strings by length instead of character by character. When one string is the beginning of the other, the shorter one is smaller.
- Treating `7//3` as 2.33. `//` is floor division.

````python
# Print each expression to check the answers live.
print('ace' in 'space' or 5%2 < 1)
print(3 == '3' or 'pen' > 'pencil')
print(7//3 > 2 or len('abs') == 3 or not '12' == 12)
print(not True and True and True)
````

````python
# Break each expression into smaller pieces.
print('ace' in 'space', 5 % 2, 5 % 2 < 1)
print(3 == '3', 'pen' > 'pencil')
print(7 // 3, 7 // 3 > 2, len('abs') == 3, '12' == 12, not '12' == 12)
print(not True, (not True) and True and True)
````

## Question 2 | Binary and Hexadecimal Representation (8 pts)

Complete the table below by converting each given number into the other two number systems: decimal, binary, and hexadecimal.

| Decimal | Binary | Hexadecimal |
|---:|---:|---:|
| | `0b101011` | |
| | | `0x5A` |

### Answers

| Decimal | Binary | Hexadecimal |
|---:|---:|---:|
| **43** | `0b101011` (given) | **`0x2B`** |
| **90** | **`0b1011010`** | `0x5A` (given) |

`0x2b` and `0x2B` are the same value; Python prints lowercase. A leading zero, as in `0b01011010`, does not change the value either.

### Explanation

**Row 1: binary → decimal**

```text
place value:  32  16   8   4   2   1
bit:           1   0   1   0   1   1
              32 +     8 +     2 + 1  = 43
```

Each binary position is worth twice the one to its right: 1, 2, 4, 8, 16, 32, counting from the right. Add the place values where the bit is 1: 32 + 8 + 2 + 1 = 43.

**Row 1: binary → hexadecimal**

```text
101011  →  10 | 1011  →  0010 | 1011  →  2 | B  →  0x2B
```

One hex digit is exactly 4 bits. Group the bits in fours **starting from the right**, pad the left group with zeros, and convert each group: 0010 is 2, and 1011 is 11, which is B. So the answer is `0x2B`.

**Row 2: hexadecimal → decimal**

In hex, A means 10, B means 11, and so on up to F, which means 15. `0x5A` is 5 × 16 + 10 = 80 + 10 = 90. The cheat sheet confirms this: `hex(90)` gives `0x5A`.

**Row 2: hexadecimal → binary**

```text
0x5A  →  5 | A  →  0101 | 1010  →  0b1011010
```

Turn each hex digit into 4 bits: 5 is 0101 and A (10) is 1010. Together that's 0101 1010; dropping the leading zero gives `0b1011010`. Check: 64 + 16 + 8 + 2 = 90.

| Hex digit | 0–9 | A | B | C | D | E | F |
|---|---|---|---|---|---|---|---|
| Value | 0–9 | 10 | 11 | 12 | 13 | 14 | 15 |

### Common mistakes

- Grouping bits from the left (`1010 | 11`), which gives the wrong answer `0xA3`.
- Forgetting that A means 10.
- Starting the place values from the left. They start at 1 on the right.

````python
# A literal written with 0b (binary) or 0x (hex) is just a number to Python.
print(0b101011)   # 43
# hex() and bin() turn a number into hex/binary text (see the cheat sheet).
# Python prints hex digits in lowercase: 0x2b is the same as 0x2B.
print(hex(43))    # 0x2b
print(0x5A)       # 90
print(bin(90))    # 0b1011010
````

Start by writing the index table:

| char | P | R | O | G | R | A | M |
|---|---|---|---|---|---|---|---|
| index | 0 | 1 | 2 | 3 | 4 | 5 | 6 |

### Program 4

```python
count = 0

for i in range(1, 13):
    if count > 4:
        break

    if i % 2 == 0 or i % 3 == 0:
        print(i)
        count += 1

print("End")
```

**Output**

```text
2
3
4
6
8
End
```

| `i` | `count` at start | `count > 4`? | even or multiple of 3? | printed | `count` after |
|---:|---:|---|---|---|---:|
| 1 | 0 | no | no | — | 0 |
| 2 | 0 | no | yes | 2 | 1 |
| 3 | 1 | no | yes | 3 | 2 |
| 4 | 2 | no | yes | 4 | 3 |
| 5 | 3 | no | no | — | 3 |
| 6 | 3 | no | yes | 6 | 4 |
| 7 | 4 | no | no | — | 4 |
| 8 | 4 | no | yes | 8 | 5 |
| 9 | 5 | **yes → `break`** | — | — | 5 |

The program prints numbers that are even **or** multiples of 3. The key detail is **where** the `break` check is: at the **top** of the loop. After 8 is printed, `count` becomes 5, but the loop doesn't stop yet. It stops at the start of the next round, when `i` is 9. So 9 is never printed, even though it is a multiple of 3. Finally, `End` is printed after the loop.

**Common mistakes**

- Printing 9 as well, which ignores the `break` at the start of the next round.
- Stopping after four numbers. `count > 4` is only True once `count` is 5.
- Printing 6 twice. It meets both conditions, but `or` only needs one side to be True, so it is printed once.

````python
# Program 4, exactly as in the exam.
count = 0

for i in range(1, 13):
    if count > 4:
        break

    if i % 2 == 0 or i % 3 == 0:
        print(i)
        count += 1

print("End")
````

## Question 4 | Predict the Output (10 pts)

```python
N = aString

i, j = 0, 0
a = 0

while i < len(N):

    j = i + 1
    while j < len(N) and N[j] == N[i]:
        j = j + 1

    if j != i + 1:
        a = a + 1

    i = j

print(a)
```

Fill in the table with the expected output for each value of `aString`.

### Answers

| `aString` | Expected output |
|---|---:|
| `'abc'` | **0** |
| `'aa'` | **1** |
| `'aabbcc'` | **3** |
| `'aaaabccdeefffffgg'` | **5** |

### Explanation

**Step 1: give every variable a job.** This is the same skeleton as Exercise 1 in Recitation 04/05. `i` marks the **start of a group** of equal neighbouring letters. `j` is a scout: it starts at `i + 1` and moves right as long as it sees the same letter. When the inner loop stops, `j` is at the start of the next group, and `i = j` jumps there.

**Step 2: what does `j != i + 1` mean?** If a group has only one letter, `j` never moves, so `j == i + 1`. If `j != i + 1`, then `j` moved at least once, so the group has **two or more** letters. So `a` counts **how many groups have length at least 2**, that is, how many groups of repeated letters there are.

**Step 3: split each string into groups.**

```text
'abc'                →  a | b | c                              →  0 groups of length ≥ 2
'aa'                 →  aa                                     →  1
'aabbcc'             →  aa | bb | cc                           →  3
'aaaabccdeefffffgg'  →  aaaa | b | cc | d | ee | fffff | gg    →  5  (aaaa, cc, ee, fffff, gg)
```

Trace for `'aabbcc'`:

| `i` | `N[i]` | `j` after the inner loop | `j != i + 1`? | `a` |
|---:|---|---:|---|---:|
| 0 | `a` | 2 | yes | 1 |
| 2 | `b` | 4 | yes | 2 |
| 4 | `c` | 6 | yes | 3 |

### Common mistakes

- Counting all groups, which gives 7 for the last string, instead of only the groups of length 2 or more.
- Counting repeated letters instead of groups, for example counting `fffff` as 4 or 5. Each group adds only 1.

````python
# Run the Question 4 program for each value in the table.
# (This checking loop uses a list only for verification; lists were not allowed in the exam answers.)
for aString in ['abc', 'aa', 'aabbcc', 'aaaabccdeefffffgg']:
    N = aString

    i, j = 0, 0
    a = 0

    while i < len(N):

        j = i + 1
        while j < len(N) and N[j] == N[i]:
            j = j + 1

        if j != i + 1:
            a = a + 1

        i = j

    print(aString, "->", a)
````

## Question 5 | Character Classifier (10 pts)

Write a Python program that reads a single character and classifies it as follows:

- If it is an English letter:
    - Display whether it is *uppercase* or *lowercase*.
    - Display whether it is a *vowel* (a, e, i, o and u) or a *consonant*.
- If it is a digit from 0 through 9, display whether it is *even* or *odd*.
- Otherwise, print *Other character*.

*Hint: Use `'A' <= letter <= 'Z'` to check whether letter is an uppercase letter; similarly for lowercase letters and digits.*

**Example Output** (the value after each prompt is the user's input):

```text
Example 1:
Enter a character: E
Uppercase vowel

Example 2:
Enter a character: b
Lowercase consonant

Example 3:
Enter a character: 7
Odd digit

Example 4:
Enter a character: #
Other character
```

### Approach

**Step 1: read the input as text.** `input()` always returns a string. Do **not** convert it with `int()` at the start: if the user types a letter, `int('E')` crashes with a `ValueError`. Convert only after you know the character is a digit.

**Step 2: decide the category with chained comparisons.** Following the hint, `'A' <= ch <= 'Z'` is True exactly for uppercase letters, because letters have consecutive code numbers. In the same way, `'a' <= ch <= 'z'` checks lowercase letters and `'0' <= ch <= '9'` checks digits. Anything else is "Other character". An `if`/`elif`/`elif`/`else` chain gives exactly these four cases.

**Step 3: vowel or consonant.** Use the membership operator: `ch in 'AEIOU'`. `in` is an operator, not a string method, so it is allowed. In the uppercase branch, compare with the uppercase vowels; in the lowercase branch, compare with the lowercase vowels.

**Step 4: even or odd.** At this point `ch` is known to be a digit, so `int(ch)` is safe. If `int(ch) % 2 == 0`, the digit is even; otherwise it is odd. Following the examples, print `Even digit` or `Odd digit`.

### Solution

````python
# Read one character. input() returns a string, so we do not convert it yet.
ch = input("Enter a character: ")

# Uppercase letters: 'A' <= ch <= 'Z' means 'A' <= ch and ch <= 'Z'.
if 'A' <= ch <= 'Z':
    # `in` checks whether ch appears in the string of vowels (an operator, not a method).
    if ch in 'AEIOU':
        print("Uppercase vowel")
    else:
        print("Uppercase consonant")
# Lowercase letters.
elif 'a' <= ch <= 'z':
    if ch in 'aeiou':
        print("Lowercase vowel")
    else:
        print("Lowercase consonant")
# Digits. Only now is int(ch) safe, because we know ch is '0' to '9'.
elif '0' <= ch <= '9':
    if int(ch) % 2 == 0:
        print("Even digit")
    else:
        print("Odd digit")
# Everything else: symbols, spaces, and so on.
else:
    print("Other character")
````

**Try it:** run the cell above with `E`, `b`, `7`, and `#`, then with `a`, `Z`, and `0`.

| Input | Output |
|---|---|
| `E` | `Uppercase vowel` |
| `b` | `Lowercase consonant` |
| `7` | `Odd digit` |
| `#` | `Other character` |
| `a` | `Lowercase vowel` |
| `Z` | `Uppercase consonant` |
| `0` | `Even digit` |

**Assumption:** the user types exactly one character, as the question says. If you are unsure about something in an exam, write your assumption down, as the cover instructions say.

### Another correct structure

You can also check "is it a letter?" first with `'A' <= letter <= 'Z' or 'a' <= letter <= 'z'`. Inside that branch you then make two separate decisions: uppercase or lowercase, and vowel or consonant (using `letter in "aeiouAEIOU"`). That needs fewer branches. If you do this, make sure the two words still appear on one line, as in the examples (`Uppercase vowel`).

### Common mistakes

- **`if ch == 'a' or 'e' or 'i' ...`** This is always True! The right side of `or`, `'e'`, is a non-empty string, which counts as True. Write `ch in 'aeiou'`, or write out every comparison as `ch == ...`.
- Converting with `int(input(...))` first. The program then crashes when the user types a letter.
- Checking uppercase letters against the lowercase vowels `'aeiou'`, so that `E` is reported as a consonant.
- Using string methods such as `ch.isupper()` or `ch.isdigit()`. They were not allowed in the exam.

## Question 6 | Saving Goal (10 pts)

Write a program that asks for an initial deposit, an annual interest rate (in %), and a target amount. Each year the balance grows by the interest rate.

The balance of next year can be calculated as:

`balance * (1 + rate / 100)`

Print the balance at the end of each year (2 decimal places) until the balance reaches or exceeds the target, then print how many years were needed.

**Example Output** (the value after each prompt is the user's input):

```text
Initial deposit: 1000
Annual interest rate (%): 4
Target amount: 1200
Year 1: 1040.00
Year 2: 1081.60
Year 3: 1124.86
Year 4: 1169.86
Year 5: 1216.65
Years needed: 5
```

### Approach

**Step 1: inputs.** Money and interest rates can have decimals, such as 3.5 percent, so use `float(input(...))`.

**Step 2: state.** You need two variables: `balance`, which starts at the deposit, and `years`, a counter that starts at 0.

**Step 3: which loop?** You don't know in advance how many years it will take, so this is a `while` loop. "Until the balance reaches or exceeds the target" means the loop stops when `balance >= target`. In other words, it **keeps going while `balance < target`**.

**Step 4: the loop body, in the right order.** First update the balance with the formula, then add one to `years`, then print. If you print before updating, the first line wrongly shows the starting deposit.

**Step 5: after the loop.** When the loop ends, `years` holds the answer. Print it once, outside the loop.

### Solution

````python
# Money and rates may have decimals, so convert with float().
deposit = float(input("Initial deposit: "))
rate = float(input("Annual interest rate (%): "))
target = float(input("Target amount: "))

# State: the current balance, and a counter for the years.
balance = deposit
years = 0

# Keep going while the target is not reached yet ("until balance >= target").
while balance < target:
    # Apply one year of interest, using the formula from the question.
    balance = balance * (1 + rate / 100)
    # Count this year.
    years += 1
    # :.2f shows exactly two decimal places.
    print(f"Year {years}: {balance:.2f}")

# Printed once, after the loop.
print("Years needed:", years)
````


**Edge cases:**

- If the deposit is already at least the target, the loop never runs and the program prints `Years needed: 0`.
- If the rate or the deposit is 0, the balance never grows and the loop never ends. We assume a positive deposit and a positive rate.

### Common mistakes

- Writing `while balance <= target`. When the balance exactly equals the target, this adds an extra year.
- Printing before updating, so that the first line shows `Year 1: 1000.00`.
- Reading the rate with `int(input(...))`, which crashes on an input like `3.5`.
- Rounding the stored balance each year with `round()`. The errors add up: Year 4 becomes `1169.85` and Year 5 becomes `1216.64`, which no longer match the example. Round only when printing, with `:.2f`.

## Question 7 | Vending Machine (15 pts)

A vending machine only accepts 1, 5, 10 and 20 yuan bills/coins. Write a Python program that:

- asks for the item price (you may assume it is a positive integer, in yuan);
- repeatedly asks the user to insert money until the total paid is at least the price. Reject any value that is not 1, 5, 10 or 20 by printing *Rejected: &lt;value&gt;*. After each accepted payment, print the progress;
- returns the change using **as few 10, 5 and 1 yuan pieces as possible**, printing only the denominations that are actually used. If there is no change, print *No change. Enjoy your drink!*

**Example Output** (the value after each prompt is the user's input):

```text
Example 1:
Item price (yuan): 7
Insert money (1, 5, 10 or 20): 3
Rejected: 3
Insert money (1, 5, 10 or 20): 20
Paid: 20 / 7
Change: 13
10 yuan x 1
1 yuan x 3
Enjoy your drink!
```

```text
Example 2:
Item price (yuan): 8
Insert money (1, 5, 10 or 20): 5
Paid: 5 / 8
Insert money (1, 5, 10 or 20): 2
Rejected: 2
Insert money (1, 5, 10 or 20): 1
Paid: 6 / 8
Insert money (1, 5, 10 or 20): 1
Paid: 7 / 8
Insert money (1, 5, 10 or 20): 1
Paid: 8 / 8
No change. Enjoy your drink!
```

### Approach

Split the problem into three parts and solve them one by one: read the price, collect the money, and give the change.

**Part A: read the price.** The price is a positive integer, so use `price = int(input(...))`. Also start an accumulator, `paid = 0`.

**Part B: collect money.** You don't know how many times the user will insert money, so use `while paid < price`. In each round, read one value. If it is 1, 5, 10, or 20, add it to `paid` and print `Paid: paid / price`. Otherwise print `Rejected: value`, and leave `paid` unchanged. Lists are not allowed, so check the four values with `or`: `money == 1 or money == 5 or money == 10 or money == 20`.

**Part C: give change.** `change = paid - price`. If it is 0, print the "No change" message. Otherwise, use the largest piece first. `change // 10` is the number of 10-yuan pieces, and `change % 10` is what is left. Do the same with 5; what remains is the number of 1-yuan pieces. Print a line only when the count is more than 0.

**Think about it: why is a 20-yuan piece never needed for change?** Before the last payment, `paid` was still less than the price, and the last payment is at most 20. So the change is at most 19.

**Think about it: why does "largest piece first" give the fewest pieces?** A 10 is two 5s, and a 5 is five 1s. Replacing smaller pieces with one bigger piece always saves pieces, so use as many big pieces as possible.

### Solution

````python
# Part A: the price is a positive integer.
price = int(input("Item price (yuan): "))
# Accumulator: the total money accepted so far.
paid = 0

# Part B: we don't know how many coins will come, so use while.
while paid < price:
    money = int(input("Insert money (1, 5, 10 or 20): "))
    # Lists are not allowed, so compare with each accepted value using `or`.
    if money == 1 or money == 5 or money == 10 or money == 20:
        paid += money
        print(f"Paid: {paid} / {price}")
    else:
        # A rejected value is NOT added to paid.
        print(f"Rejected: {money}")

# Part C: change.
change = paid - price

if change == 0:
    print("No change. Enjoy your drink!")
else:
    print("Change:", change)
    # Largest piece first: // gives how many pieces, % gives what is left.
    tens = change // 10
    change = change % 10
    fives = change // 5
    ones = change % 5
    # Print only the denominations that are actually used.
    if tens > 0:
        print(f"10 yuan x {tens}")
    if fives > 0:
        print(f"5 yuan x {fives}")
    if ones > 0:
        print(f"1 yuan x {ones}")
    print("Enjoy your drink!")
````

**Try it:** run the cell with `7`, `3`, `20` (Example 1) and then with `8`, `5`, `2`, `1`, `1`, `1` (Example 2). The output should match the examples above exactly.

**Extra test:** price `1`, insert `20` → `Change: 19`, `10 yuan x 1`, `5 yuan x 1`, `1 yuan x 4`, `Enjoy your drink!`.

### Common mistakes

- **`if money == 1 or 5 or 10 or 20`** This is always True! `5` by itself counts as True. Write every comparison out in full as `money == ...`.
- `money in [1, 5, 10, 20]` uses a list, which was not allowed in the exam.
- Adding rejected money to `paid`.
- Printing denominations that are not used, such as `5 yuan x 0`.
- Forgetting `int()`. The string `'5'` never equals the number `5`, so every payment is rejected and the loop never ends.
- Forgetting `Enjoy your drink!` when there is change.

## Question 8 | Prime Finder (15 pts)

Write a Python program that reads two integers `start` and `end` (you may assume `start ≤ end`) and prints all prime numbers in that range (inclusive) on one line, separated by spaces, followed by the total count.

A prime is an integer greater than 1 whose only divisors are 1 and itself.

**Example Output** (the value after each prompt is the user's input):

```text
Example 1:
Start: 10
End: 30
11 13 17 19 23 29
Total primes: 6
```

```text
Example 2:
Start: 1
End: 10
2 3 5 7
Total primes: 4
```

### Approach

**Step 1: two levels of repetition.** You repeat for every number in the range, and for each number you try many divisors. Two levels of repetition means nested loops: the outer loop goes through the numbers, and the inner loop tests one number.

**Step 2: the outer loop.** "Inclusive" means `end` must be included, so use `range(start, end + 1)`. Remember that the stop value of `range` is not included.

**Step 3: test one number with a flag.** Numbers less than or equal to 1 are not prime, so skip them. For any other number, start by assuming it is prime: `is_prime = True`. Then try every divisor from 2 to `number - 1`. As soon as one divides evenly, set `is_prime = False` and `break`, because a single divisor proves the number is not prime. After the inner loop, `is_prime` holds the answer.

**Step 4: print on one line.** For each prime, `print(number, end=" ")` keeps the numbers on one line, and `count += 1` counts it. After the outer loop, an empty `print()` ends that line. Then print the total.

### Solution

````python
start = int(input("Start: "))
end = int(input("End: "))

# Counter for the primes found.
count = 0

# "Inclusive" means end is included, so the stop value is end + 1.
for number in range(start, end + 1):
    # Primes are greater than 1, so 1, 0, and negative numbers are skipped.
    if number > 1:
        # Flag: assume it is prime until a divisor proves otherwise.
        # Reset it for EVERY number.
        is_prime = True
        for divisor in range(2, number):
            if number % divisor == 0:
                is_prime = False
                # One divisor is enough: stop testing this number.
                break
        if is_prime:
            # end=" " keeps all primes on one line, separated by spaces.
            print(number, end=" ")
            count += 1

# Finish the line of primes, then print the total.
print()
print("Total primes:", count)
````

**Try it:** run the cell with `10`, `30`, and then with `1`, `10`.

```text
Start: 10
End: 30
11 13 17 19 23 29
Total primes: 6
```

```text
Start: 1
End: 10
2 3 5 7
Total primes: 4
```

Notice the `1` in Example 2: 1 is not prime, which is why the `number > 1` check is needed. Also, `print(number, end=" ")` leaves one invisible space at the end of the line. That is normally fine. To avoid it, print a space **before** every prime except the first one.

**Think about it: how can the inner loop be made faster?** If a number greater than 1 is not prime, it has a divisor no larger than its square root. So you only need to test divisors up to the square root, for example with `range(2, int(number ** 0.5) + 1)`. For exam-sized inputs, `range(2, number)` is also fine.

### Common mistakes

- Using `range(start, end)`, which misses `end` itself (for example, 29 when `end` is 29).
- Treating 1 as prime.
- Writing `is_prime = True` before the outer loop instead of inside it. After the first non-prime, every later number is reported as not prime.
- Printing inside the inner loop, so that a number is printed several times.
- Forgetting the final `print()`, so that `Total primes` appears on the same line as the primes.
