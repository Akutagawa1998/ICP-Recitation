## Review 1 — interactive evaluation, scripts, and visible output

The interactive prompt immediately shows the value of a bare expression, which makes it useful for quick calculations. A script executes the same expression but does not automatically display its value. Use `print(...)` when the program should produce visible output.

Python applies multiplication before addition, just as in ordinary arithmetic; parentheses make the intended grouping explicit.

```python
review_value = 2 + 3 * 4
print("Calculated value:", review_value)

script_example = "5 + 7"
print("A bare expression can run in a script without producing visible output:", script_example)
```

## Review 2 — syntax, strings, and safe error inspection

Python must parse source code before it can execute it. Text that violates the language grammar raises `SyntaxError`. Quotation marks instead tell Python that text is a string literal. If an unquoted word has not been defined as a name, evaluating it fails at runtime.

For a recitation notebook, expected failures should be caught so later cells can still run.

```python
print("A different string example")

example_source = "3 4"
try:
    compile(example_source, "<review-example>", "exec")
except SyntaxError as error:
    print("Caught safely:", type(error).__name__)
```

## Review 3 — one integer, several representations

Decimal, binary, and hexadecimal are different written representations of an integer. Python binary literals begin with `0b`; hexadecimal literals begin with `0x`. Evaluating either literal gives an integer, while `bin(...)` and `hex(...)` return strings that show a chosen representation.

```python
review_number = 26
print("decimal:", review_number)
print("binary:", bin(review_number))
print("hexadecimal:", hex(review_number))
assert int(bin(review_number), 2) == review_number
assert int(hex(review_number), 16) == review_number
```
