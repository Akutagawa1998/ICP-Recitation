## 热身 A：先看运算符，再看类型

整数通常是 `int`，带小数点的数通常是 `float`。`/` 做真正的除法，即使两个整数能整除，结果也为 `float`。`//` 向负无穷取整；这与 `int()` 把小数部分朝 0 去掉不同。对这里的非零除数，余数满足 `a == (a // b) * b + (a % b)`；结果不为 0 时，余数与除数同号。

先算括号，再算幂，再算乘除、整除、取余，最后加减。同一优先级的乘除取余从左往右。不要只写数字而忘了类型：一项出现浮点数可能改变最终类型。下面是热身例子，数值与原题不同。

````python
print(29 / 6, type(29 / 6).__name__)
print(29 // 6, 29 % 6)
print(-29 // 6, int(-29 / 6))  # 向下取整 vs 朝 0 截断
print(4 + 6 * 2, (4 + 6) * 2)
````

## 热身 B：值、名字与类型

`"8"` 是字符串，`8` 是整数。字符串的 `+` 连接字符串；数值的 `+` 做加法。Python 不会替你猜测应该把哪一边转换。`int()`、`float()` 把合适的字符串变成数；`str()` 把值变成字符串。字符串重复需要整数次数。

变量名不用引号；有引号时得到的是那几个字符本身。`x = y` 把当时的 `y` 值赋给 `x`，不会建立以后自动同步的公式。多重赋值先求出整个右边的旧值，再按左边名字的顺序绑定。

````python
text_count = "8"
print(text_count + "2", int(text_count) + 2)
print("la" * 3)
x = 9
y = x
x = 2
print(x, y)
x, y = y, x  # 先得到旧的右边，再赋给左边
print(x, y)
````

## 热身 C：字符串边界与输出边界

字符串开头的引号必须由同类引号结束；另一种引号可以作为内容。`\'` 和 `\"` 表示引号字符，`\n` 是换行，`\t` 是制表符。制表符移动到下一个制表位，显示距离受当前位置和环境影响，**不是固定几个空格**。

`print(value1, value2, sep=..., end=...)` 先把值变成文本，再在相邻参数之间放 `sep`，最后加 `end`。默认 `sep=' '`、`end='\n'`；字符串里面原有的空格不会被删除。`repr()` 是观察工具：它用带引号、带转义的形式描述文本，这些外层引号不是原始 `print` 自动打印的内容。

````python
print("red", "blue", sep=" / ", end="!\n")
message = "X\tY\nZ"
print(message)
print("辅助 repr：", repr(message))
````

## 热身 D：格式化是生成显示文本

f-string 的 `{表达式:格式说明}` 先计算值，再生成字符串。字段宽度是**最小**宽度，内容太长通常不会被剪掉。`<` 左对齐，`>` 右对齐；紧挨对齐符之前的字符可以作为填充。

精度取决于被格式化的类型：浮点数的 `.2f` 表示小数点后两位；浮点数只写 `.2`、省略 `f` 时，使用类似通用格式的有效数字规则，不等于固定两位小数。字符串的 `.2` 最多保留两个字符。先读值的类型，再解释精度；不要把宽度当精度。格式化不改变原变量的数值。

````python
amount = 6.789
print("|" + f"{amount:9.2f}" + "|")
print("|" + f"{'owl':.>7}" + "|")
print("|" + f"{'sparrow':<7.2}" + "|")
print(amount)  # 原来的值没有变
````

### 1.3 `(34 // 10) * (1.0 + 2)`

预测：结果 = ___；类型 = ___。

````python
value = 23 % 4.0
print(value, type(value).__name__)
````

**逐项讲解：** `23 = 5 * 4.0 + 3.0`，结果 `3.0`，类型为 `float`。余数即使没有非零小数部分，也不会因此自动变回 int。

### 1.6 `25 + 30 / 6`

预测：结果 = ___；类型 = ___。

````python
value = 25 + 30 / 6
print(value, type(value).__name__)
````

**逐项讲解：** 先算 `30 / 6 = 5.0`，再算 `25 + 5.0 = 30.0`，类型为 `float`。不能把 `25 + 30` 先加起来。

### 1.7 `-7 % 3`

预测：结果 = ___；类型 = ___。

````python
value = -7 % 3
print(value, type(value).__name__)
````

**逐项讲解：** 先求向下取整的商：`-7 // 3 = -3`，不是朝 0 截断的 `-2`。再由 `-7 = (-3) * 3 + 2` 得到余数 `2`，类型为 `int`。这不是把正数余数直接加上负号。

### 1.8 `100 - 25 * 3 % 4`

预测：结果 = ___；类型 = ___。

````python
value = 100 - 25 * 3 % 4
print(value, type(value).__name__)
````

**逐项讲解：** 乘法与取余同级，先从左到右算 `25 * 3 = 75`，再算 `75 % 4 = 3`，最后 `100 - 3 = 97`，类型为 `int`。易错：先做减法，或者把 `%` 当成更低优先级。

### 本题考察什么

这题检查输入转换、数学分式的程序表达、乘法符号必须显式写出，以及输出单位。你需要保留公式中的比例关系与偏移量，不能改变运算的组合。

**怎么做：** 先读下面恢复排版的公式，再用可编辑输入完成换算，并用容易心算的温度检验。

**恢复原分式的排版：**

$$F = \frac{9}{5}C + 32$$

其中 $C$ 是摄氏温度，$F$ 是华氏温度。这不是 `9 / (5*C + 32)`；写代码前先确认分数线的范围。

````python
celsius_text = "25"  # 交互版：celsius_text = input("摄氏温度：")
celsius = float(celsius_text)
fahrenheit = (9 / 5) * celsius + 32
print(f"{celsius:.1f} °C = {fahrenheit:.1f} °F")
````

**推导与易错点：** 先把 `9/5` 作为比例，乘摄氏温度，再加 32；在 Python 中相邻的 `(9 / 5)` 和 `celsius` 之间必须写 `*`。`/` 会得到浮点比值 1.8，不能用 `//` 代替。示例 25 °C 对应 77 °F。可检查 0 °C → 32 °F、100 °C → 212 °F、−40 °C → −40 °F；这三个值同时帮助发现漏加 32、比例写反、误用整除等错误。
