## 热身 A：先看运算符，再看类型

整数通常是 `int`，带小数点的数通常是 `float`。`/` 做真正的除法，即使两个整数能整除，结果也为 `float`。`//` 向负无穷取整；这与 `int()` 把小数部分朝 0 去掉不同。对这里的非零除数，余数满足 `a == (a // b) * b + (a % b)`；结果不为 0 时，余数与除数同号。

先算括号，再算幂，再算乘除、整除、取余，最后加减。同一优先级的乘除取余从左往右。不要只写数字而忘了类型：一项出现浮点数可能改变最终类型。下面是热身例子，数值与原题不同。

> **English translation:**
> ## Warm-up A: Read the Operators, Then the Types
>
> Integers are usually `int` values, and numbers containing a decimal point are usually `float` values. `/` (read “slash,” or “divided by” in arithmetic) performs true division; even when two integers divide evenly, the result is a `float`. `//` (read “double slash,” or “floor-divided by”) rounds the quotient toward negative infinity. This differs from `int()`, which removes the fractional part toward zero. For the nonzero divisors used here, the remainder satisfies `a == (a // b) * b + (a % b)`: read `==` as “is equal to,” `*` as “times,” `+` as “plus,” and `%` as “modulo.” When the remainder is nonzero, it has the same sign as the divisor.
>
> Evaluate parentheses first, then exponentiation, then multiplication, true division, floor division, and modulo, and finally addition and subtraction. Operators at the same multiplication/division level are evaluated from left to right. Do not report only the number and forget its type: the presence of a floating-point value may change the final type. The next cell is a warm-up example whose values differ from the original exercise.


````python
print(29 / 6, type(29 / 6).__name__)
print(29 // 6, 29 % 6)
print(-29 // 6, int(-29 / 6))  # 向下取整 vs 朝 0 截断
print(4 + 6 * 2, (4 + 6) * 2)
````

## 热身 B：值、名字与类型

`"8"` 是字符串，`8` 是整数。字符串的 `+` 连接字符串；数值的 `+` 做加法。Python 不会替你猜测应该把哪一边转换。`int()`、`float()` 把合适的字符串变成数；`str()` 把值变成字符串。字符串重复需要整数次数。

变量名不用引号；有引号时得到的是那几个字符本身。`x = y` 把当时的 `y` 值赋给 `x`，不会建立以后自动同步的公式。多重赋值先求出整个右边的旧值，再按左边名字的顺序绑定。

> **English translation:**
> ## Warm-up B: Values, Names, and Types
>
> `"8"` is a string, whereas `8` is an integer. For strings, `+` (read “plus”) concatenates; for numbers, it adds. Python does not guess which operand you intended to convert. `int()` and `float()` convert suitable strings to numbers; `str()` converts a value to a string. String repetition requires an integer repetition count.
>
> A variable name is written without quotation marks. With quotes, the result is those literal characters. In `x = y`, read `=` as “is assigned” or “gets”: the current value of `y` is assigned to `x`; this does not create a formula that automatically stays synchronized. Multiple assignment first evaluates all old values on the right-hand side and then binds them to the names on the left in order.


````python
text_count = "8"
print(text_count + "2", int(text_count) + 2)
print("la" * 3)
x = 9
y = x
x = 2
print(x, y)
x, y = y, x  # 先得到旧的右边，再赋给左边
print(x, y)
````

## 热身 C：字符串边界与输出边界

字符串开头的引号必须由同类引号结束；另一种引号可以作为内容。`\'` 和 `\"` 表示引号字符，`\n` 是换行，`\t` 是制表符。制表符移动到下一个制表位，显示距离受当前位置和环境影响，**不是固定几个空格**。

`print(value1, value2, sep=..., end=...)` 先把值变成文本，再在相邻参数之间放 `sep`，最后加 `end`。默认 `sep=' '`、`end='\n'`；字符串里面原有的空格不会被删除。`repr()` 是观察工具：它用带引号、带转义的形式描述文本，这些外层引号不是原始 `print` 自动打印的内容。

> **English translation:**
> ## Warm-up C: String Boundaries and Output Boundaries
>
> The quote that starts a string must be closed by the same kind of quote; the other kind may appear as content. `\'` is read “backslash single quote,” `\"` is “backslash double quote,” `\n` is “backslash n” (a newline), and `\t` is “backslash t” (a tab). A tab advances to the next tab stop, so its displayed width depends on the current position and environment; it is **not a fixed number of spaces**.
>
> `print(value1, value2, sep=..., end=...)` first converts values to text, places `sep` between adjacent arguments, and appends `end`. Read the commas as “comma,” `=` as assignment in these keyword arguments, and `...` as “ellipsis.” The defaults are `sep=' '` and `end='\n'`; spaces already inside a string are not removed. `repr()` is an inspection tool: it describes text with surrounding quotes and escape sequences. Those outer quotes are not characters automatically printed by the original `print` call.


````python
print("red", "blue", sep=" / ", end="!\n")
message = "X\tY\nZ"
print(message)
print("辅助 repr：", repr(message))
````

## 热身 D：格式化是生成显示文本

f-string 的 `{表达式:格式说明}` 先计算值，再生成字符串。字段宽度是**最小**宽度，内容太长通常不会被剪掉。`<` 左对齐，`>` 右对齐；紧挨对齐符之前的字符可以作为填充。

精度取决于被格式化的类型：浮点数的 `.2f` 表示小数点后两位；浮点数只写 `.2`、省略 `f` 时，使用类似通用格式的有效数字规则，不等于固定两位小数。字符串的 `.2` 最多保留两个字符。先读值的类型，再解释精度；不要把宽度当精度。格式化不改变原变量的数值。

> **English translation:**
> ## Warm-up D: Formatting Produces Display Text
>
> In an f-string, `{expression:format_specifier}` first evaluates the expression and then produces a string. Read `{` and `}` as “left/right curly brace” and `:` as “colon.” Field width is a **minimum** width; content that is too long is normally not cut off. `<` is the left-alignment marker and `>` is the right-alignment marker. A character immediately before an alignment marker can be used as the fill character.
>
> Precision depends on the formatted value's type. For a float, `.2f` (read “dot two f”) means two digits after the decimal point. If a float uses `.2` without `f`, Python applies significant-digit rules similar to general formatting; it does not mean exactly two decimal places. For a string, `.2` keeps at most two characters. Read the value's type before interpreting precision, and do not confuse width with precision. Formatting does not change the original variable's numeric value.


````python
amount = 6.789
print("|" + f"{amount:9.2f}" + "|")
print("|" + f"{'owl':.>7}" + "|")
print("|" + f"{'sparrow':<7.2}" + "|")
print(amount)  # 原来的值没有变
````

## 热身 E：输入 → 处理 → 输出

`input()` 总是返回字符串。先按意义选 `int()`（整数数量）或 `float()`（允许小数），再计算，最后打印结果和单位。先用能心算的输入检查公式方向和比例。

> **English translation:**
> ## Warm-up E: Input → Processing → Output
>
> `input()` always returns a string. Choose `int()` for an integer quantity or `float()` when decimals are allowed, then calculate, and finally print the result with its unit. Start with an input you can calculate mentally so you can check the direction and scale of the formula. Read `→` as “goes to.”


````python
distance_text = "3.5"  # 交互时替换为 input("距离（km）：")
distance_km = float(distance_text)
distance_m = distance_km * 1000
print(f"距离：{distance_m:.1f} m")
````
