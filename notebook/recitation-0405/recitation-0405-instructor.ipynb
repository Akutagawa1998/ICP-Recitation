## Exercise 1 | Predict the output (W)

Trace this source program and fill the expected-output column for each input.

```python
N = input('Enter a string: ')
i, j = 0, 0
while i < len(N):
    j = i + 1
    while j < len(N) and N[j] == N[i]:
        j = j + 1
    print(N[i], end='')
    i = j
```

| Input | Expected output |
| --- | --- |
| `abcdeabcdeabcde` | |
| `aabbccddeeff` | |
| `abcaabbccaaabbbcccddd` | |

Write the exact concatenated output. The shown code does not add separators or a final newline.

After predicting, run the code cell below once for each input in the table and type that input at the prompt.


````python
# This function does the same work as the source program, so we can check our predictions.
# 这个函数和原题程序做的是同一件事，用来核对我们的预测。
# Instead of printing letter by letter, it builds a string and returns it,
# so the result is easy to print or test.
# 它不逐个打印字母，而是拼出一个字符串再返回，方便打印或检验。
def keep_one_character_per_run(text):
    # result collects one letter from each "run" (a group of equal letters next to each other).
    # result 收集每一"段"（相邻且相同的一组字母）里保留下来的那一个字母。
    result = ""
    # i marks the start of the current run. String indexes start at 0.
    # i 指向当前这一段的第一个字母。字符串下标从 0 开始。
    i = 0
    # len(text) is the number of characters. Keep going while i is still inside the string.
    # len(text) 是字符个数。只要 i 还在字符串范围内，就继续。
    while i < len(text):
        # j is a "scout": it starts at the position right after i.
        # j 是"探路"的指针：从 i 的下一个位置出发。
        j = i + 1
        # Move j to the right while it sees the same letter as text[i].
        # 只要 j 看到的字母和 text[i] 相同，就继续右移。
        # j < len(text) must come FIRST: when it is False, Python does not evaluate text[j]
        # (short-circuit evaluation), so we never read past the end of the string.
        # 必须先写 j < len(text)：它为假时 Python 不再计算 text[j]（短路求值），所以不会越界。
        while j < len(text) and text[j] == text[i]:
            j = j + 1
        # Keep only one letter for the whole run.
        # 整段只保留一个字母。
        result = result + text[i]
        # j is now the first different letter, so jump straight to the next run.
        # 现在 j 指向第一个不同的字母，所以直接跳到下一段的开头。
        i = j
    # return sends the finished string back to the caller.
    # return 把拼好的字符串交回给调用它的地方。
    return result

# input() shows the prompt, waits for the user, and always returns a string.
# input() 显示提示语并等待用户输入，返回值永远是字符串。
# Run this cell once for each input in the table.
# 表格里有几个输入，就把这个单元格运行几次。
text = input("Enter a string: ")
# Call the function and print the result, e.g. "aabbccddeeff" -> "abcdef".
# 调用函数并打印结果，例如 "aabbccddeeff" → "abcdef"。
print(keep_one_character_per_run(text))
````

## Exercise 2 | Calories Burned (W/F)

Running on a particular treadmill, you burn **4.2 calories per minute**. Write a program that uses a loop to display the number of calories burned after **10, 15, 20, 25, and 30 minutes**.

Show both a `while`-loop solution and a `for`-loop solution. The source does not specify exact labels or numeric formatting; use clear labels and one decimal place.


````python
# A named variable makes the formula easier to read than the bare number 4.2.
# 用有名字的变量代替直接写 4.2，公式更容易读懂。
calories_per_minute = 4.2
# Start state: the first time point is 10 minutes.
# 初始状态：第一个时间点是 10 分钟。
minute = 10
# Stop condition: <= 30 so that 30 itself is included.
# 停止条件：用 <= 30，保证 30 本身也会输出。
while minute <= 30:
    # Each value comes straight from the minute, so no running total is needed.
    # 每个值都能由分钟数直接算出，不需要累加。
    calories = minute * calories_per_minute
    # f-string: values inside {} are inserted into the text; :.1f shows one decimal place.
    # f-string：{} 里的值会被填进文字中；:.1f 表示保留一位小数。
    print(f"After {minute} minutes: {calories:.1f} calories")
    # Update: go to the next time point. Without this line, the loop would never end.
    # 更新：前进到下一个时间点。少了这一行，循环就永远停不下来。
    minute = minute + 5
````

````python
# Same rate as in the while version.
# 与 while 版本相同的消耗速度。
calories_per_minute = 4.2
# range(start, stop, step) produces 10, 15, 20, 25, 30.
# range(起点, 终点, 步长) 产生 10, 15, 20, 25, 30。
# The stop value is NOT included, so we write 31 to keep 30.
# 终点不包含在内，所以写 31 才能包含 30。
# This one line replaces the start, the condition, and the update of the while version.
# 这一行代替了 while 版本里的起点、条件和更新三部分。
for minute in range(10, 31, 5):
    # Same calculation as before: minutes times calories per minute.
    # 和前面一样：分钟数乘以每分钟的消耗。
    calories = minute * calories_per_minute
    # :.1f keeps one decimal place, e.g. 42.0.
    # :.1f 保留一位小数，例如 42.0。
    print(f"After {minute} minutes: {calories:.1f} calories")
````

## Exercise 3 | Fibonacci (W/F)

Write a program that asks the user for a strictly positive integer `n` and displays the Fibonacci sequence up to the `n`-th value. The source defines:

- `F_0 = 0`
- `F_1 = 1`
- for `n > 1`, `F_n = F_(n-1) + F_(n-2)`

**Interpretation:** because the source indexes from zero, display `F_0` through `F_n`, inclusive. Assume the user enters a valid positive integer. Show both `while` and `for` versions.


````python
# Return a list with F_0 through F_n (n + 1 numbers), built with a while loop.
# 返回包含 F_0 到 F_n（共 n + 1 个数）的列表，用 while 循环构造。
def fibonacci_through_while(n):
    # The definition gives F_0 = 0 and F_1 = 1, so the list starts with them.
    # A list remembers earlier values; each new value needs the two values before it.
    # 定义直接给出 F_0 = 0 和 F_1 = 1，所以列表一开始就放入它们。
    # 用列表记住之前的值，因为每个新值都要用到它前面的两个值。
    sequence = [0, 1]
    # Index 2 is the first value we must calculate.
    # 下标 2 是第一个需要计算的值。
    index = 2
    # Keep calculating until F_n is done.
    # 一直算到 F_n 为止。
    while index <= n:
        # F_index = F_(index-1) + F_(index-2): add the two previous stored values.
        # F_index = F_(index-1) + F_(index-2)：把列表中前两个已保存的值相加。
        next_value = sequence[index - 1] + sequence[index - 2]
        # append() adds a value to the end of a list.
        # append() 把一个值加到列表末尾。
        sequence.append(next_value)
        # Update, so the loop moves forward and eventually stops.
        # 更新下标，让循环前进并最终停止。
        index = index + 1
    # The slice [:n + 1] keeps the items at indexes 0 to n.
    # For n = 1 the loop never runs, and this still returns [0, 1].
    # 切片 [:n + 1] 保留下标 0 到 n 的元素。
    # n = 1 时循环一次也不执行，返回的仍然是 [0, 1]。
    return sequence[:n + 1]

# input() returns text such as "8"; int() converts it to the number 8,
# so that we can compare it (index <= n) and do arithmetic (n + 1).
# input() 返回的是文本，例如 "8"；int() 把它转成整数 8，
# 这样才能比较大小（index <= n）和做运算（n + 1）。
n = int(input("Enter a strictly positive integer n: "))
# Show the whole list, e.g. 8 -> [0, 1, 1, 2, 3, 5, 8, 13, 21].
# 打印整个列表，例如输入 8 → [0, 1, 1, 2, 3, 5, 8, 13, 21]。
print(fibonacci_through_while(n))
````

````python
# Same task as the while version, written with a for loop.
# 与 while 版本任务相同，改用 for 循环。
def fibonacci_through_for(n):
    # Start with F_0 and F_1, which the definition gives us.
    # 从定义给出的 F_0 和 F_1 开始。
    sequence = [0, 1]
    # range(2, n + 1) gives index = 2, 3, ..., n (the stop value n + 1 is not included).
    # range takes care of the start, the stop, and the update for us.
    # range(2, n + 1) 依次给出 index = 2, 3, ..., n（终点 n + 1 不包含）。
    # 起点、终点和更新都由 range 负责。
    for index in range(2, n + 1):
        # Add the two previous values to get the next one.
        # 前两个值相加得到下一个值。
        next_value = sequence[index - 1] + sequence[index - 2]
        # Store it at the end of the list.
        # 把它存到列表末尾。
        sequence.append(next_value)
    # Keep exactly F_0 ... F_n.
    # 只保留 F_0 到 F_n。
    return sequence[:n + 1]

# Read n as text and convert it to an integer.
# 读入文本形式的 n，再转换成整数。
n = int(input("Enter a strictly positive integer n: "))
# The result should match the while version for the same input.
# 相同输入下，结果应与 while 版本完全一样。
print(fibonacci_through_for(n))
````

## Exercise 4 | Guessing Competition (W/F)

Write a Python number-guessing game. The computer randomly chooses an integer from **1 through 100**, inclusive. The player has at most **7 guesses**.

After each guess:

- if it is too high, display `"Too high!"`;
- if it is too low, display `"Too low!"`;
- if it is correct, display `"Correct!"` and the number of guesses used.

After every incorrect guess, display the current inclusive possible range. After a seventh incorrect guess, show the updated range, then display `"Game over!"` and reveal the secret number.

Show both `while` and `for` versions. Each attempt asks the player for one guess. Assume every guess is a valid integer.


````python
# The random module provides random numbers. We must import it before using random.randint.
# random 模块提供随机数；使用 random.randint 之前必须先导入它。
import random

# One game of "guess the number", written with a while loop.
# We use a function so that `return` can end the whole game right after a correct guess.
# 用 while 循环实现的一局猜数字游戏。
# 写成函数，是为了猜中后能用 return 立刻结束整局游戏。
def play_guessing_game_while():
    # randint(a, b) returns a random integer from a to b. BOTH ends are included (unlike range).
    # randint(a, b) 返回 a 到 b 之间的随机整数，两端都包含（这一点和 range 不同）。
    secret_number = random.randint(1, 100)
    # The range where the secret can still be. It shrinks after every wrong guess.
    # 秘密数仍可能所在的范围，每猜错一次就缩小一次。
    lowest_possible = 1
    highest_possible = 100
    # How many guesses have been used so far.
    # 目前已经猜了几次。
    attempt_number = 0
    # The player has at most 7 guesses.
    # 玩家最多猜 7 次。
    while attempt_number < 7:
        # Update the counter FIRST: the prompt then shows 1, 2, ..., 7,
        # and we cannot forget the update.
        # 先更新计数：提示会依次显示 1, 2, ..., 7，也不会忘记更新。
        attempt_number = attempt_number + 1
        # input() returns text; int() turns it into a whole number we can compare.
        # input() 返回文本；int() 把它变成可以比较大小的整数。
        guess = int(input(f"Guess {attempt_number}: "))
        if guess == secret_number:
            # Messages required by the source.
            # 原题要求的提示信息。
            print("Correct!")
            print("Guesses used:", attempt_number)
            # return leaves the function at once: the loop stops and "Game over!" is skipped.
            # return 立即离开函数：循环随之结束，也不会打印 "Game over!"。
            return
        if guess > secret_number:
            print("Too high!")
            # Too high, so the secret is at most guess - 1 (the guess itself is wrong).
            # min(a, b) returns the smaller value. It keeps the bound from growing
            # if the player guesses outside the current range (e.g. range 1-24, guess 40).
            # 猜大了，所以秘密数最多是 guess - 1（猜过的数本身不可能）。
            # min(a, b) 返回较小的值；即使玩家猜了范围外的数（如范围 1-24 却猜 40），上界也不会变大。
            highest_possible = min(highest_possible, guess - 1)
        else:
            print("Too low!")
            # Too low, so the secret is at least guess + 1.
            # max(a, b) returns the larger value, for the same reason.
            # 猜小了，所以秘密数至少是 guess + 1。
            # max(a, b) 返回较大的值，道理同上。
            lowest_possible = max(lowest_possible, guess + 1)
        # Show the updated range after every wrong guess.
        # 每次猜错后显示更新后的范围。
        print(f"Possible range: {lowest_possible}-{highest_possible}")
    # We only get here if all 7 guesses were wrong.
    # 只有 7 次全部猜错，才会执行到这里。
    print("Game over!")
    print("Secret number:", secret_number)

# Call the function to start a game. The secret number is different on every run.
# 调用函数开始一局游戏；每次运行秘密数都不同。
play_guessing_game_while()
````

````python
# Each cell can run on its own, so this cell imports random again.
# 每个单元格都可以单独运行，所以这里再导入一次 random。
import random

# The same game, written with a for loop.
# 同一个游戏，改用 for 循环。
def play_guessing_game_for():
    # A random secret from 1 to 100; randint includes both ends.
    # 1 到 100 之间的随机秘密数；randint 两端都包含。
    secret_number = random.randint(1, 100)
    # The range where the secret can still be.
    # 秘密数仍可能所在的范围。
    lowest_possible = 1
    highest_possible = 100
    # range(1, 8) gives the attempt numbers 1, 2, ..., 7 directly,
    # so the for version needs no counter of its own.
    # range(1, 8) 直接给出尝试编号 1, 2, ..., 7，所以 for 版本不需要自己维护计数器。
    for attempt_number in range(1, 8):
        # Read one guess and convert it to an integer.
        # 读入一次猜测并转成整数。
        guess = int(input(f"Guess {attempt_number}: "))
        if guess == secret_number:
            print("Correct!")
            print("Guesses used:", attempt_number)
            # return also ends a for loop early, because it leaves the whole function.
            # return 会离开整个函数，所以 for 循环也提前结束。
            return
        if guess > secret_number:
            print("Too high!")
            # Lower the upper bound; min() never lets it grow.
            # 降低上界；min() 保证上界不会变大。
            highest_possible = min(highest_possible, guess - 1)
        else:
            print("Too low!")
            # Raise the lower bound; max() never lets it shrink.
            # 提高下界；max() 保证下界不会变小。
            lowest_possible = max(lowest_possible, guess + 1)
        print(f"Possible range: {lowest_possible}-{highest_possible}")
    # Reached only after 7 wrong guesses.
    # 只有 7 次都猜错才会执行到这里。
    print("Game over!")
    print("Secret number:", secret_number)

# Start a game.
# 开始一局游戏。
play_guessing_game_for()
````

## Exercise 5 | Repetition of a substring (W)

Write a program that outputs the number of times one string appears in another.

Source examples:

```text
Write a string: abcdeabcdeabcde
Write another string: abc
String 'abc' appears 3 times.

Write a string: aaabcaa
Write another string: aa
String 'aa' appears 2 times.
```

The source says “appears into another,” which likely means “appears in another.” The second example establishes **non-overlapping**, case-sensitive matches. Assume the substring is not empty. Use a `while` loop.


````python
# Count how many times substring appears in text: non-overlapping and case-sensitive.
# 统计 substring 在 text 中出现的次数：不重叠，区分大小写。
def count_non_overlapping(text, substring):
    # Counter for the matches found so far.
    # 计数器：目前找到的匹配次数。
    count = 0
    # index is where the "window" starts. The window is a piece of text as long as substring.
    # index 是"窗口"的起点；窗口是一段和 substring 一样长的文字。
    index = 0
    # The last start where a whole window still fits inside text.
    # 窗口还能完整放进 text 的最后一个起点。
    last_start = len(text) - len(substring)
    # We use while (not for) because the step changes:
    # 1 after a miss, len(substring) after a match. A for loop over range has a fixed step.
    # Use <= so that the last possible start is also checked.
    # 这里用 while 而不是 for，因为步长会变：没匹配前进 1，匹配了前进 len(substring)；
    # for ... in range 的步长是固定的。用 <= 是为了把最后一个可能的起点也检查到。
    while index <= last_start:
        # Slicing text[a:b] takes the characters from index a up to, but not including, b.
        # 切片 text[a:b] 取下标 a 到 b 之前（不含 b）的字符。
        if text[index:index + len(substring)] == substring:
            count = count + 1
            # Non-overlapping: after a match, jump past the whole match.
            # ("aaabcaa" with "aa" gives 2, as in the source example, not 3.)
            # 不重叠：匹配成功后跳过整个匹配。
            # （所以 "aaabcaa" 中 "aa" 数到 2，和原题样例一致，而不是 3。）
            index = index + len(substring)
        else:
            # No match here: move one character and try again.
            # 这里不匹配：前进一个字符再试。
            index = index + 1
    # Give the total back to the caller.
    # 把总数交回给调用者。
    return count

# Both inputs are text already, so no int() is needed. The prompts follow the source.
# 两个输入本来就是文本，不需要 int()。提示语沿用原题。
text = input("Write a string: ")
substring = input("Write another string: ")
# Store the answer in a clearly named variable before printing it.
# 先把结果存进一个名字清楚的变量，再打印。
number_of_matches = count_non_overlapping(text, substring)
# The sentence format comes from the source example.
# 句子格式来自原题样例。
print(f"String '{substring}' appears {number_of_matches} times.")
````

## Exercise 6 | Greatest Common Divisor (W)

The greatest common divisor of two positive integers, `n` and `m`, is the largest integer `d` that divides both evenly. Use this source algorithm:

1. Initialize `d` to the smaller of `m` and `n`.
2. While `d` does not evenly divide `m` **or** `d` does not evenly divide `n`, decrease `d` by 1.
3. Report `d`.

Write a program that reads two positive integers and follows this exact decrementing-divisor algorithm. The PDF bullet layout does not indent the decrement clearly; treat it as the loop body. Assume the user enters two valid positive integers.


````python
# Greatest common divisor, following exactly the algorithm given in the exercise.
# 最大公约数，严格按照题目给出的算法。
def greatest_common_divisor(first_number, second_number):
    # min() returns the smaller number. A common divisor can never be larger than it.
    # min() 返回较小的那个数；公约数不可能比它还大。
    divisor = min(first_number, second_number)
    # % gives the remainder of a division; remainder 0 means "divides evenly".
    # Keep going while divisor fails for AT LEAST ONE number. That is why we use `or`:
    # with `and`, 84 and 30 would stop at 30, which divides only 30.
    # % 求除法的余数；余数为 0 表示"能整除"。
    # 只要 divisor 对"至少一个"数不能整除，就继续往下找，所以用 `or`；
    # 如果写成 `and`，84 和 30 会在 30 就停下，而 30 只能整除 30。
    while first_number % divisor != 0 or second_number % divisor != 0:
        # Try the next smaller candidate. The loop always stops, because 1 divides every number.
        # 试下一个更小的候选数。循环一定会停，因为 1 能整除任何数。
        divisor = divisor - 1
    # We count down, so the first common divisor we meet is the greatest one.
    # 我们是从大往小找的，所以遇到的第一个公约数就是最大的。
    return divisor

# The source says to read two positive integers from the user; int() converts the text.
# 原题要求从用户处读取两个正整数；int() 把输入的文本转成整数。
first_number = int(input("Enter the first positive integer: "))
second_number = int(input("Enter the second positive integer: "))
# Example: 84 and 30 -> 6.
# 例如：84 和 30 → 6。
print("Greatest common divisor:", greatest_common_divisor(first_number, second_number))
````

````python
# Draw a triangle of stars whose rows get shorter: rows, rows - 1, ..., 1 stars.
# 画一个逐行变短的星号三角形：每行依次是 rows、rows - 1、…、1 颗星。
def draw_decreasing_stars(rows):
    # Outer loop: one row per round. range(rows, 0, -1) counts down from rows to 1
    # (0 is not included), and that number is exactly how many stars this row needs.
    # 外层循环：每轮负责一行。range(rows, 0, -1) 从 rows 倒数到 1（不包含 0），
    # 这个数正好就是这一行需要的星号数。
    for number_of_stars in range(rows, 0, -1):
        # `_` is a name for a loop variable we do not use; we only want to repeat.
        # `_` 表示"用不到的循环变量"，这里只需要重复固定的次数。
        for _ in range(number_of_stars):
            # end="" keeps the next star on the same line.
            # end="" 让下一颗星留在同一行。
            print("*", end="")
        # The row is finished: move to a new line.
        # 这一行画完了：换行。
        print()

# The source asks for the number of rows; int() turns the typed text into a number.
# 原题要求输入行数；int() 把输入的文本转成整数。
rows = int(input("Enter the number of rows: "))
draw_decreasing_stars(rows)
````

````python
# Draw rows of the form "#" + spaces + "#", with one more space on each row.
# 画出形如 "#" + 若干空格 + "#" 的行，每一行多一个空格。
def draw_widening_hashes(rows):
    # row_index = 0, 1, 2, ...; it is also the number of spaces between the two hashes.
    # row_index 依次为 0, 1, 2, ...；它同时也是两个 # 之间的空格数。
    for row_index in range(rows):
        # Left hash; end="" stays on the same line.
        # 左边的 #；end="" 让光标留在同一行。
        print("#", end="")
        # row_index spaces. Row 0 gets no spaces, so it prints "##".
        # 打印 row_index 个空格；第 0 行没有空格，所以输出 "##"。
        for _ in range(row_index):
            print(" ", end="")
        # Right hash. No end="" here, so print also moves to a new line.
        # 右边的 #。这里没有 end=""，所以 print 打印后顺便换行。
        print("#")

# Read the number of rows from the user.
# 从用户处读取行数。
rows = int(input("Enter the number of rows: "))
draw_widening_hashes(rows)
````

````python
# Draw a right-aligned triangle of letters. The letters continue from row to row
# and wrap from Z back to A.
# 画一个靠右对齐的字母三角形；字母跨行连续，Z 之后回到 A。
def draw_letter_triangle(rows):
    # How many letters have been printed so far. It is created ONCE, before the outer loop,
    # because the letters continue across rows (row 2 starts with B, not A).
    # 已经打印了多少个字母。它只在外层循环之前创建一次，
    # 因为字母要跨行连续（第 2 行从 B 开始，而不是 A）。
    letter_index = 0
    # row_number = 1, 2, ..., rows. Row number r contains r letters.
    # row_number 依次为 1, 2, ..., rows；第 r 行有 r 个字母。
    for row_number in range(1, rows + 1):
        # Leading spaces: rows - row_number (with 8 rows: 7 on row 1, then 6, ..., 0 on row 8).
        # 前导空格数是 rows - row_number（8 行时：第 1 行 7 个，然后 6 个……第 8 行 0 个）。
        for _ in range(rows - row_number):
            print(" ", end="")
        # Print the letters of this row.
        # 打印这一行的字母。
        for column_index in range(row_number):
            # Put a space BEFORE every letter except the first one,
            # so no row ends with an extra space.
            # 除第一个字母外，在每个字母"前面"放一个空格，这样行尾不会多出空格。
            if column_index > 0:
                print(" ", end="")
            # ord() and chr() convert between a character and its code number:
            #     ord("A") -> 65,  ord("B") -> 66,  ...      chr(65) -> "A",  chr(67) -> "C"
            # So chr(ord("A") + k) is the letter k places after "A" (k = 0 -> "A", k = 2 -> "C").
            # % is calculated before +, so this is ord("A") + (letter_index % 26).
            # % 26 wraps the count back to 0 after "Z": 26 -> "A", 27 -> "B".
            # ord() 和 chr() 在"字符"和它的"编号"之间互相转换：
            #     ord("A") → 65，ord("B") → 66……      chr(65) → "A"，chr(67) → "C"
            # 所以 chr(ord("A") + k) 就是 "A" 之后第 k 个字母（k = 0 → "A"，k = 2 → "C"）。
            # % 比 + 先计算，所以这里是 ord("A") + (letter_index % 26)。
            # % 26 让编号在 "Z" 之后回到 0：26 → "A"，27 → "B"。
            # If students have not seen chr/ord, string indexing gives the same letter:
            #     alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
            #     letter = alphabet[letter_index % 26]
            # 如果学生还没学过 chr/ord，用字符串下标也能取到同一个字母：
            #     alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
            #     letter = alphabet[letter_index % 26]
            letter = chr(ord("A") + letter_index % 26)
            print(letter, end="")
            # Move on to the next letter. This counter is never reset.
            # 前进到下一个字母；这个计数器永远不重置。
            letter_index = letter_index + 1
        # The row is finished.
        # 这一行结束，换行。
        print()

# Read the number of rows from the user (the source example uses 8).
# 从用户处读取行数（原题样例是 8）。
rows = int(input("Enter the number of rows: "))
draw_letter_triangle(rows)
````

````python
# Draw number_of_triangles triangles, each with triangle_size rows.
# Three loops, three jobs: which triangle, which row, how many stars in that row.
# 画 number_of_triangles 个三角形，每个有 triangle_size 行。
# 三层循环、三个职责：第几个三角形、第几行、这一行有几颗星。
def draw_triangles(number_of_triangles, triangle_size):
    # Job 1: repeat the whole triangle. We do not need the triangle's number, so we use `_`.
    # 职责 1：重复画整个三角形。用不到三角形的编号，所以用 `_`。
    for _ in range(number_of_triangles):
        # Job 2: the rows of one triangle, row_number = 1, 2, ..., triangle_size.
        # 职责 2：一个三角形的各行，row_number = 1, 2, ..., triangle_size。
        for row_number in range(1, triangle_size + 1):
            # Job 3: row number r has r stars.
            # 职责 3：第 r 行有 r 颗星。
            for _ in range(row_number):
                print("*", end="")
            # End the row. No extra print() between triangles, so there is no blank line.
            # 换行。三角形之间不额外 print()，所以不会出现空行。
            print()

# Two inputs, with the same prompts as the source example.
# 两个输入，提示语和原题样例相同。
number_of_triangles = int(input("Enter the number of triangles: "))
triangle_size = int(input("Enter the size of one triangle: "))
draw_triangles(number_of_triangles, triangle_size)
````

## Exercise 9 | Dice statics (F)

Write a program that rolls a six-sided die **1,000 times**. Count how many ones, twos, threes, fours, fives, and sixes were rolled, then display all six counts.

The source title says **“Dice statics,”** which is likely a typo for **“Dice statistics.”** The source does not specify labels or a random seed. Use clear labels and a fixed local seed for a reproducible demonstration.


````python
# The random module provides random numbers.
# random 模块提供随机数。
import random

# Roll a six-sided die number_of_rolls times and return how often each face appeared.
# 掷六面骰子 number_of_rolls 次，返回每个点数各出现了几次。
def roll_die_counts(number_of_rolls, seed):
    # random.Random(seed) creates our own random-number generator. With a fixed seed it
    # produces the same "random" sequence on every run, so the class always sees the same counts.
    # random.Random(seed) 创建一个属于我们自己的随机数生成器。种子固定时，
    # 每次运行都会产生相同的"随机"序列，所以全班每次看到的计数都一样。
    randomizer = random.Random(seed)
    # One counter per face, kept in a list instead of six separate variables.
    # Index 0 is for face 1, ..., index 5 is for face 6.
    # 每个点数一个计数器，放在一个列表里，而不是写六个变量。
    # 下标 0 对应点数 1……下标 5 对应点数 6。
    counts = [0, 0, 0, 0, 0, 0]
    # The number of rolls is known in advance, so a for loop fits. `_`: the loop variable is unused.
    # 掷骰次数事先已知，所以用 for 循环。`_` 表示用不到循环变量。
    for _ in range(number_of_rolls):
        # randint(1, 6) includes both 1 and 6.
        # randint(1, 6) 两端都包含，结果是 1 到 6 之一。
        face = randomizer.randint(1, 6)
        # Faces are 1-6 but list indexes are 0-5, so subtract 1.
        # (counts[face] would fail for face 6, because index 6 does not exist.)
        # 点数是 1–6，列表下标是 0–5，所以要减 1。
        # （写成 counts[face] 的话，点数 6 会访问不存在的下标 6 而报错。）
        counts[face - 1] = counts[face - 1] + 1
    # Return the whole list of six counts.
    # 返回包含 6 个计数的整个列表。
    return counts

# 1,000 rolls with seed 409. Whatever the seed, the six counts always add up to 1,000.
# 用种子 409 掷 1000 次。不管种子是多少，6 个计数之和一定是 1000。
counts = roll_die_counts(1000, 409)
# range(1, 7) gives the faces 1 to 6 (7 is not included).
# range(1, 7) 产生点数 1 到 6（不包含 7）。
for face in range(1, 7):
    print(f"{face}: {counts[face - 1]}")
````

## Exercise 10 | Theme Park Ride Admission (W/F)

Ten people are waiting for a roller coaster with only five available seats. For each person, ask for a height in centimeters.

- A person must be at least **120 cm** tall to ride.
- If a person is too short, display `"Sorry, you are not tall enough."` and skip to the next person.
- If a person is tall enough, allow that person to board and display `"Welcome aboard!"` as shown in the source example.
- Once five people have boarded, stop checking the remaining people.
- Display `"The ride is full!"` only if five people board.
- Always display the final passenger count.

Source example:

```text
Person 1 - Enter height: 135
Welcome aboard!
Person 2 - Enter height: 110
Sorry, you are not tall enough.
...
The ride is full!
5 passengers are on board.
```

Show both `while` and `for` versions. Assume every entered height is a valid whole number of centimeters.


````python
# Check up to 10 people and fill at most 5 seats, using a while loop.
# 用 while 循环：最多检查 10 个人，最多坐满 5 个座位。
def admit_passengers_while():
    # Two different numbers: passengers on board (max 5) and people checked (max 10).
    # 两个不同的数：已上车人数（最多 5）和已检查人数（最多 10）。
    passenger_count = 0
    person_number = 0
    # Both stop reasons are in the condition: 10 people checked, or 5 seats taken.
    # 两个停止理由都写在条件里：已检查 10 人，或者 5 个座位坐满。
    while person_number < 10 and passenger_count < 5:
        # Update FIRST, before any continue. If this line came after continue,
        # a short person would skip it and the program would ask the same person again and again.
        # 一定要先更新，放在 continue 之前。如果这一行写在 continue 后面，
        # 身高不够的人会跳过它，程序就会一遍又一遍地询问同一个人。
        person_number = person_number + 1
        # The prompt follows the source example: "Person 1 - Enter height: ".
        # 提示语沿用原题样例："Person 1 - Enter height: "。
        height = int(input(f"Person {person_number} - Enter height: "))
        # "At least 120 cm" means 120 is allowed, so we reject only heights below 120.
        # "至少 120 cm"包含 120，所以只拒绝低于 120 的身高。
        if height < 120:
            print("Sorry, you are not tall enough.")
            # continue skips the boarding steps below and goes back to the loop condition.
            # continue 跳过下面的上车步骤，回到循环条件。
            continue
        print("Welcome aboard!")
        # Only tall-enough people take a seat.
        # 只有身高合格的人才占座位。
        passenger_count = passenger_count + 1
    # The loop may also end after 10 people with fewer than 5 passengers,
    # so check before saying the ride is full.
    # 循环也可能在检查完 10 人、但不足 5 人时结束，所以先判断再说"满员"。
    if passenger_count == 5:
        print("The ride is full!")
    # The final count is always shown.
    # 最终人数总是要显示。
    print(f"{passenger_count} passengers are on board.")

# Start checking the queue.
# 开始逐个检查排队的人。
admit_passengers_while()
````

````python
# The same rules, using a for loop.
# 同样的规则，改用 for 循环。
def admit_passengers_for():
    # Passengers on board so far.
    # 目前已上车的人数。
    passenger_count = 0
    # range(1, 11) gives person_number = 1, 2, ..., 10 and moves to the next person for us,
    # so continue is safe here.
    # range(1, 11) 依次给出 person_number = 1, 2, ..., 10，并自动前进到下一个人，
    # 所以这里用 continue 是安全的。
    for person_number in range(1, 11):
        height = int(input(f"Person {person_number} - Enter height: "))
        if height < 120:
            print("Sorry, you are not tall enough.")
            # Skip to the next person; this person takes no seat.
            # 跳到下一个人；这个人不占座位。
            continue
        print("Welcome aboard!")
        passenger_count = passenger_count + 1
        # range only knows about the 10 people, not the 5 seats,
        # so we need an explicit break when the ride is full.
        # range 只知道 10 个人，不知道 5 个座位，所以坐满时需要显式 break。
        if passenger_count == 5:
            # break ends the loop: the remaining people are not asked.
            # break 结束循环：剩下的人不会再被询问。
            break
    # Same ending as the while version.
    # 结尾与 while 版本相同。
    if passenger_count == 5:
        print("The ride is full!")
    print(f"{passenger_count} passengers are on board.")

# Start checking the queue.
# 开始逐个检查排队的人。
admit_passengers_for()
````

## Exercise 11 | Password Generator (F)

Repeatedly ask the user for a password length. If the user enters `0`, terminate. If the user enters a positive value `X`, create a password containing exactly `X` random numerical digits from `0` through `9`.

Source example (the displayed passwords are illustrative random output, not fixed answers):

```text
Password size? > 4
9605
Password size? > 9
654395954
Password size? > 1
4
Password size? > 0
```

A password is a string, so a leading zero must be preserved.

The source labels this exercise **F**, but the number of requests is not known in advance. This notebook therefore repeats the requests with `while True` and stops with `break` when the user enters `0`. The required `for` loop generates exactly `X` digits for each password. Assume every entered size is a valid non-negative integer.


````python
# The random module provides random digits.
# random 模块用来生成随机数字。
import random

# Keep making passwords until the user enters 0.
# 不断生成密码，直到用户输入 0。
def generate_passwords():
    # We do not know in advance how many requests will come, so the outer loop is
    # `while True` ("repeat forever"); the real stopping rule is the break below.
    # (The source labels this exercise F; the required for loop builds the digits.)
    # 事先不知道用户会请求几次，所以外层用 while True（"一直重复"），
    # 真正的停止规则是下面的 break。（原题标注 F；题目要求的 for 循环用来生成每一位数字。）
    while True:
        # Read the requested length and convert it to an integer.
        # 读入要求的密码长度，并转成整数。
        password_size = int(input("Password size? > "))
        # 0 is a sentinel: a special value that means "stop".
        # 0 是终止值：一个表示"停止"的特殊值。
        if password_size == 0:
            # break (not continue) ends the whole loop, so the program finishes.
            # 用 break（不是 continue）结束整个循环，程序随之结束。
            break
        # Build the password as a STRING: a leading 0 is kept ("0427"), and + joins text.
        # 密码要用"字符串"来拼：开头的 0 能保留（如 "0427"），而且字符串的 + 是拼接。
        password = ""
        # Inner for loop: repeat exactly password_size times, one digit each time.
        # 内层 for 循环：正好重复 password_size 次，每次生成一位。
        for _ in range(password_size):
            # randint(0, 9) includes both 0 and 9, so every digit is possible.
            # randint(0, 9) 两端都包含，所以每个数字都可能出现。
            digit = random.randint(0, 9)
            # str() turns the number into text so that + adds it to the end of the password.
            # str() 把数字变成文本，这样 + 才是把它接到密码末尾。
            password = password + str(digit)
        # Show this password, then ask for the next size.
        # 显示这个密码，然后询问下一个长度。
        print(password)

# Start the generator.
# 启动密码生成器。
generate_passwords()
````

## Recap

- A `while` loop is controlled by a condition and must update the state that can end it.
- A `for` loop is useful when iterating over a known range or sequence.
- Nested loops solve a repeated task inside another repeated task.
- `continue` skips the rest of the current iteration; `break` ends the nearest loop.
- Trace tables, boundary cases, and fixed random seeds make loop behavior easier to verify.
