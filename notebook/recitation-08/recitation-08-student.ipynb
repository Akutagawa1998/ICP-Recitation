## Exercise 2: Total Sales

Design a program that asks the user to enter a store's sales for each day of the week. The amounts should be stored in a list. Use a loop to calculate the total sales for the week and display the result.

Collect seven numeric amounts, then use a second loop to add them. The function below uses real input and returns the stored list after displaying the total.

````python
def collect_weekly_sales():
    sales = []
    for day in range(1, 8):
        amount = float(input("Sales for day " + str(day) + ": "))
        sales.append(amount)
    total = 0.0
    for amount in sales:
        total = total + amount
    print("Weekly total: " + format(total, ".2f"))
    return sales
````

### Demonstration

These seven daily amounts are added classroom examples: `120.0, 135.5, 98.0, 150.0, 210.0, 300.0, 175.0`. Their total is `1188.50`. `patch` supplies these responses only during this demonstration and restores normal input afterward.

````python
from unittest.mock import patch

sales_answers = iter(["120.0", "135.5", "98.0", "150.0", "210.0", "300.0", "175.0"])
def sales_demo_input(prompt):
    answer = next(sales_answers)
    print(prompt + answer)
    return answer

with patch("builtins.input", sales_demo_input):
    weekly_sales = collect_weekly_sales()
print("Stored sales:", weekly_sales)
````

## Exercise 5: Larger Than n

In a program, write a function that accepts two arguments: a list, and a number n. Assume that the list contains numbers. The function should display all of the numbers in the list that are greater than the number n.

Use `>` so a value equal to `n` is excluded. The first call displays `15` and `23`; the no-match and empty-list calls display nothing.

````python
def display_larger_than(numbers, n):
    for number in numbers:
        if number > n:
            print(number)

display_larger_than([4, 10, 15, 10, 23, -2], 10)
display_larger_than([4, 10, -2], 10)
display_larger_than([], 10)
````

## Exercise 6: Lo Shu Magic Square

The Lo Shu Magic Square is a grid with 3 rows and 3 columns. It has the following properties:

- The grid contains the numbers 1 through 9 exactly.
- The sum of each row, each column, and each diagonal all add up to the same number.

The source figure is:

```text
4 9 2
3 5 7
8 1 6
```

Each row, column, and diagonal in this figure sums to 15.

In a program you can simulate a magic square using a two-dimensional list. Write a function that accepts a two-dimensional list as an argument and determines whether the list is a Lo Shu Magic Square. Test the function in a program.

Interpret "1 through 9 exactly" as each number appearing once. Their total is 45, so the common row sum must be 15. Check the size, the nine values, all rows, all columns, and both diagonals. The input is assumed to be a two-dimensional list of numbers. The added negative examples isolate rules that must not be skipped.

````python
def is_lo_shu(grid):
    if len(grid) != 3:
        return False
    for row in grid:
        if len(row) != 3:
            return False
    values = []
    for row in grid:
        for number in row:
            values.append(number)
    if sorted(values) != list(range(1, 10)):
        return False
    for row in grid:
        if sum(row) != 15:
            return False
    for column in range(3):
        column_total = grid[0][column] + grid[1][column] + grid[2][column]
        if column_total != 15:
            return False
    if grid[0][0] + grid[1][1] + grid[2][2] != 15:
        return False
    if grid[0][2] + grid[1][1] + grid[2][0] != 15:
        return False
    return True

magic_square = [[4, 9, 2], [3, 5, 7], [8, 1, 6]]
print("Source square:", is_lo_shu(magic_square))
print("Repeated values:", is_lo_shu([[5, 5, 5], [5, 5, 5], [5, 5, 5]]))
print("Wrong diagonals:", is_lo_shu([[1, 5, 9], [6, 7, 2], [8, 3, 4]]))
print("Wrong size:", is_lo_shu([[4, 9, 2], [3, 5, 7]]))
print("Out of range:", is_lo_shu([[4, 9, 2], [3, 5, 7], [8, 1, 10]]))
````

This question is about data related to students enrolled in a course and their grades during a semester.

The data about students and their grades is stored in a list of lists where each sublist represents one student. All sublists have the following seven pieces of information:

student_grades = [

### 1: computeAverage(studentList)

Write a function named `computeAverage()` that takes a parameter named `studentList` in the same format as `student_grades` and returns a list of lists where each sublist represents one student. Each result sublist contains the student id, first name, last name, and average grade for all four tests, all as strings.

 ['5634423', 'Bob', 'Smith', '87.5'],

### 2: highestAverage(studentList)

Write a function `highestAverage()` that takes a parameter named `studentList` in the same format as `student_grades` and returns a list with the information of the best performing student.

Example call: `highestAverage(student_grades)`

Expected returned value: `['8754543', 'Joe', 'Public']`.

Keep IDs and names as strings. Convert each grade to a number for addition, divide by four, then convert the average back with `str`. Build a new result list, leaving the original records unchanged.

````python
student_grades = [
    ['2345243', 'John', 'Doe', '79', '89', '49', '79'],
    ['5634423', 'Bob', 'Smith', '90', '90', '80', '90'],
    ['1433145', 'Jane', 'Roe', '61', '71', '81', '81'],
    ['8754543', 'Joe', 'Public', '82', '93', '92', '92'],
    ['3142354', 'Jean', 'Martin', '95', '37', '39', '98']
]

def computeAverage(studentList):
    average_records = []
    for student in studentList:
        grade_total = 0.0
        for grade in student[3:]:
            grade_total = grade_total + float(grade)
        average = grade_total / 4
        average_record = student[:3] + [str(average)]
        average_records.append(average_record)
    return average_records

average_records = computeAverage(student_grades)
for record in average_records:
    print(record)
````

Reuse `computeAverage`, compare averages as numbers, and return only the three identity fields. The source does not define ties or an empty input: this solution keeps the first student in a tie and returns `[]` for an empty list.

````python
def highestAverage(studentList):
    averages = computeAverage(studentList)
    if len(averages) == 0:
        return []
    best_student = averages[0]
    for student in averages[1:]:
        if float(student[3]) > float(best_student[3]):
            best_student = student
    return best_student[:3]

print(highestAverage(student_grades))
print("Empty input:", highestAverage([]))
````

## Exercise 8: Random Attendance System

Write a program that stores student records in a list. Each record is a tuple containing a student's name and absence count, for example:

Stu_list = [("Alice", 0), ("Bob", 0), ("Charlie", 0)]

Write the following functions:

### 1: register_students()

Ask the user to enter all student names. Set each student's absence count to 0 and return the list of records.

### 2: take_attendance(Stu_list, n)

Randomly select `n` different students. For each selected student, display their name and ask whether they are present. If absent, increase their absence count by one. Update the original list. Assume attendance input is `"P"` for present or `"A"` for absent.

Ask for a student's name and display their absence count. If the name is not found, display `"Student not enrolled."`.

Implementation choices for details not specified in the source: enter one name per line and finish with a blank line; reject duplicate exact names; match names case-sensitively; allow `n` from 0 to the enrolled count. The parameter `student` in `check_absences(student)` means the full registered list. Invalid attendance responses are requested again. `main()` returns the final list on quit only to make its state easy to check.

### 1: Register records

Create one tuple `(name, 0)` for each unique name and place the tuples in a list. Return when the user enters a blank line.

````python
import random

def register_students():
    Stu_list = []
    while True:
        name = input("Enter a student name (blank to finish): ").strip()
        if name == "":
            return Stu_list
        enrolled = False
        for student_name, absences in Stu_list:
            if student_name == name:
                enrolled = True
        if enrolled:
            print("Student already enrolled.")
        else:
            Stu_list.append((name, 0))
````

### 2: Update the original list

Sample indices without replacement. A tuple cannot be changed item by item, but its position in the outer list can be replaced with `(name, absences + 1)`. Present students keep their counts.

````python
def take_attendance(Stu_list, n):
    if n < 0 or n > len(Stu_list):
        print("Choose a number from 0 to " + str(len(Stu_list)) + ".")
        return
    selected_indices = random.sample(range(len(Stu_list)), n)
    for index in selected_indices:
        name, absences = Stu_list[index]
        status = input("Is " + name + " present? (P/A): ").strip().upper()
        while status not in ("P", "A"):
            status = input("Please enter P or A: ").strip().upper()
        if status == "A":
            Stu_list[index] = (name, absences + 1)
````

### 3: Search records

Search the list passed as `student`. Return as soon as a matching name is found; show the missing-name message only after all records have been checked.

````python
def check_absences(student):
    name = input("Enter the student's name: ").strip()
    for student_name, absences in student:
        if student_name == name:
            print(student_name + ": " + str(absences) + " absence(s)")
            return
    print("Student not enrolled.")
````

### 4: Keep the same list between menu choices

Register once before the menu loop. Every attendance and search call uses that same list, so absence counts persist between rounds. Calling `main()` later starts a new interactive session.

````python
def main():
    print("Enroll students...")
    Stu_list = register_students()
    while True:
        print("1. Take attendance")
        print("2. Check a student's absence count")
        print("3. Quit")
        choice = input("Choose an option: ").strip()
        if choice == "1":
            try:
                n = int(input("How many students should be selected? "))
            except ValueError:
                print("Please enter an integer.")
                continue
            take_attendance(Stu_list, n)
        elif choice == "2":
            check_absences(Stu_list)
        elif choice == "3":
            return Stu_list
        else:
            print("Choose 1, 2, or 3.")
````

### Demonstration

The following cell supplies fixed responses and fixes only the two demonstration selections to match the source: Bob/Alice, then Charlie/Bob. Bob is absent twice, so the final records are `[('Alice', 0), ('Bob', 2), ('Charlie', 0)]`. David is not enrolled. Normal input and real random sampling are restored after the `with` blocks.

````python
from unittest.mock import patch

attendance_answers = iter([
    "Alice", "Bob", "Charlie", "",
    "1", "2", "A", "P",
    "2", "Bob",
    "1", "2", "P", "A",
    "2", "Bob",
    "2", "David", "3"
])
def attendance_demo_input(prompt):
    answer = next(attendance_answers)
    print(prompt + answer)
    return answer

selected_rounds = [[1, 0], [2, 1]]
with patch("builtins.input", attendance_demo_input):
    with patch("random.sample", side_effect=selected_rounds):
        final_records = main()
print("Final records:", final_records)
````
